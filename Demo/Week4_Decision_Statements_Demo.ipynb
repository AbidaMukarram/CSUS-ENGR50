## 1. The `if` Statement: Syntax Basics

Python's `if` statement has two syntax pieces that are easy to forget if you're used to another language: a **colon** at the end of the condition, and **consistent indentation** for everything inside the block. There is no `end` keyword — the block simply ends when the indentation returns to the previous level.

```python
age = 20
if age >= 18:
    print("You are an adult.")
```

If the condition is False, the indented block simply doesn't run — nothing is printed, and the program moves on:

```python
age = 15
if age >= 18:
    print("You are an adult.")
print("This line always runs, regardless of age.")
```

## 3. `if` / `elif` / `else`: Three or More Branches

This is the pattern **Programming Project 4 actually needs**. `elif` (short for "else if") lets you chain multiple conditions together. Python checks them **in order** and runs the first one that's True — everything after that is skipped, even if it would also technically be True.

```python
def temp_category(temp):
    if temp >= 90:
        return "Hot"
    elif temp >= 70:
        return "Warm"
    elif temp >= 50:
        return "Cool"
    else:
        return "Cold"

for t in [95, 75, 55, 30]:
    print(t, "->", temp_category(t))
```

Notice the structure: any number of `elif`s, and one final `else` that catches everything not matched above. The `else` is optional, but including one is good practice — it guarantees your function always returns *something*, even for a case you didn't think to test.

## 4. The Critical Gotcha: `elif` vs. Separate `if` Statements

This is the single most important thing to get right for Programming Project 4. Using several **independent** `if` statements instead of a proper `elif` chain can produce subtly wrong results — even when it looks like it should work.

```python
def bad_check(score):
    result = []
    if score >= 60:
        result.append("Pass")
    if score < 60:                 # a SEPARATE, independent if -- not elif!
        result.append("Fail")
    return result

def good_check(score):
    if score >= 60:
        return "Pass"
    else:
        return "Fail"

print(bad_check(75))    # technically still correct here...
print(good_check(75))
```

In this particular example, both versions happen to agree — but `bad_check` only works because the two conditions (`>= 60` and `< 60`) happen to be perfect opposites. The moment your conditions get more complex (like three discriminant cases that aren't simple opposites of each other), separate `if` statements become genuinely risky: it becomes possible for **more than one condition to be True at once**, or for a case to fall through **all of them** with nothing catching it. An `elif` chain guarantees exactly one branch runs, by construction — that guarantee is exactly what you want for a case-based calculation like the quadratic solver.

## 5. Worked Example: A Multi-Branch Calculation

This example mirrors the *structure* Programming Project 4 needs: compute a value first, then branch based on it — without using the quadratic formula itself.

```python
def bmi_category(weight_kg, height_m):
    bmi = weight_kg / (height_m ** 2)
    if bmi < 18.5:
        category = "Underweight"
    elif bmi < 25:
        category = "Normal"
    elif bmi < 30:
        category = "Overweight"
    else:
        category = "Obese"
    print(f"BMI = {bmi:.1f} -> {category}")

bmi_category(70, 1.75)
bmi_category(50, 1.75)
bmi_category(95, 1.75)
```

Notice the pattern: **one calculation, then a branch on the result.** This is exactly the shape of Programming Project 4 — compute the discriminant first, then branch on whether it's positive, zero, or negative.
