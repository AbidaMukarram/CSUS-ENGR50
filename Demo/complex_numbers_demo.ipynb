# Complex Numbers in Engineering — Python Demo

This notebook accompanies **Deck 1: Complex Numbers in Engineering**.  
It demonstrates how to work with complex numbers using Python's built-in `complex` type, the `cmath` module, and `numpy`.

**Topics covered:**
- Creating complex numbers
- Arithmetic (add, subtract, multiply, divide)
- Conjugate and magnitude
- Polar form and conversion
- Solving polynomial equations
- Engineering application: AC impedance

```python
import cmath
import numpy as np
import matplotlib.pyplot as plt

print('Libraries loaded. Ready to work with complex numbers!')
```

## 1. Creating Complex Numbers

In Python, `j` is the imaginary unit (engineers use `j` to avoid confusion with current `i`).

```python
# --- Method 1: Literal notation ---
z1 = 3 + 4j
z2 = 1 - 2j
z3 = 5j        # Pure imaginary
z4 = 7 + 0j    # Pure real

# --- Method 2: complex() constructor ---
z5 = complex(3, 4)   # same as 3 + 4j

print(f'z1 = {z1}')
print(f'z2 = {z2}')
print(f'z3 = {z3}')
print(f'z4 = {z4}')
print(f'z5 = {z5}')

# --- Extracting real and imaginary parts ---
print(f'\nReal part of z1:      {z1.real}')
print(f'Imaginary part of z1: {z1.imag}')
```

## 2. Arithmetic Operations

```python
z1 = 3 + 4j
z2 = 1 - 2j

# Addition
print(f'z1 + z2 = {z1 + z2}')    # (3+1) + (4-2)j = 4 + 2j

# Subtraction
print(f'z1 - z2 = {z1 - z2}')    # (3-1) + (4+2)j = 2 + 6j

# Multiplication  (FOIL: (3+4j)(1-2j))
product = z1 * z2
print(f'z1 * z2 = {product}')     # 3-6j+4j-8j² = 3-2j+8 = 11 + 2j

# Division
quotient = z1 / z2
print(f'z1 / z2 = {quotient:.4f}')  # multiply by conjugate of denominator

# Powers
print(f'j^2 = {1j**2}')           # should be -1
print(f'j^3 = {1j**3}')           # should be -j
print(f'j^4 = {1j**4}')           # should be 1
```

## 3. Conjugate and Magnitude

```python
z = 3 + 4j

# Conjugate: flip sign of imaginary part
z_conj = z.conjugate()
print(f'z        = {z}')
print(f'conj(z)  = {z_conj}')

# Magnitude (absolute value): |z| = sqrt(a² + b²)
magnitude = abs(z)              # Python built-in
magnitude_manual = (z.real**2 + z.imag**2) ** 0.5
print(f'|z| = {magnitude}')    # sqrt(9 + 16) = sqrt(25) = 5

# Verify: z * conj(z) = |z|²
print(f'z * conj(z) = {z * z_conj}')   # should be 25 + 0j

# Phase angle (argument)
phase = cmath.phase(z)
print(f'Phase = {phase:.4f} rad = {np.degrees(phase):.2f}°')  # atan(4/3) ≈ 53.13°
```

## 4. Polar Form

Any complex number can be written as $z = r \cdot e^{j\theta} = r(\cos\theta + j\sin\theta)$

```python
z = 3 + 4j

# --- Rectangular → Polar ---
r, theta = cmath.polar(z)   # returns (magnitude, phase_in_radians)
print(f'Rectangular: {z}')
print(f'Polar:  r = {r}, θ = {theta:.4f} rad ({np.degrees(theta):.2f}°)')

# --- Polar → Rectangular ---
z_back = cmath.rect(r, theta)   # converts (r, theta) → complex
print(f'Back to rectangular: {z_back:.4f}')

# --- Euler's formula: e^(jθ) ---
theta = np.pi / 4   # 45 degrees
z_euler = cmath.exp(1j * theta)
print(f'\ne^(j·π/4) = {z_euler:.4f}')
print(f'Which is:   cos(45°) + j·sin(45°) = {np.cos(theta):.4f} + {np.sin(theta):.4f}j')

# --- Visualize on the complex plane ---
fig, ax = plt.subplots(figsize=(6, 6))
ax.axhline(0, color='gray', linewidth=0.8)
ax.axvline(0, color='gray', linewidth=0.8)
ax.plot(z.real, z.imag, 'bo', markersize=10, label='z = 3+4j')
ax.annotate('z = 3+4j', xy=(z.real, z.imag), xytext=(z.real+0.3, z.imag+0.2), fontsize=12)
ax.arrow(0, 0, z.real, z.imag, head_width=0.15, head_length=0.15, fc='blue', ec='blue')
ax.set_xlim(-1, 6); ax.set_ylim(-1, 6)
ax.set_xlabel('Real'); ax.set_ylabel('Imaginary')
ax.set_title('Complex Number on the Complex Plane')
ax.grid(True, alpha=0.3)
ax.set_aspect('equal')
plt.tight_layout()
plt.show()
print(f'Magnitude |z| = {r:.1f}, Angle θ = {np.degrees(theta):.1f}°')
```

## 5. Solving Polynomial Equations

Polynomials with real coefficients may have complex roots. `numpy.roots()` finds them all.

```python
# --- Example 1: x² + 1 = 0 (roots are ±j) ---
coeffs1 = [1, 0, 1]   # 1·x² + 0·x + 1
roots1 = np.roots(coeffs1)
print('x² + 1 = 0')
print(f'Roots: {roots1}')  # [0+1j, 0-1j]

# --- Example 2: x² + 2x + 5 = 0 (complex conjugate pair) ---
# Discriminant: b² - 4ac = 4 - 20 = -16  (negative → complex roots)
coeffs2 = [1, 2, 5]
roots2 = np.roots(coeffs2)
print('\nx² + 2x + 5 = 0')
for r in roots2:
    print(f'  Root: {r:.4f}  |magnitude: {abs(r):.4f}')

# --- Example 3: Characteristic equation of an RLC circuit ---
# LCs² + Rs + 1/C = 0 → s² + (R/L)s + 1/(LC) = 0
R, L, C = 100, 0.1, 1e-6   # ohms, henries, farads
coeffs3 = [1, R/L, 1/(L*C)]
s_poles = np.roots(coeffs3)
print(f'\nRLC circuit poles (s = σ ± jω):')
for s in s_poles:
    sigma = s.real
    omega = abs(s.imag)
    print(f'  s = {s:.1f}  →  σ={sigma:.1f} (decay rate),  ω={omega:.1f} rad/s (oscillation freq)')
```

## 6. Engineering Application: AC Circuit Impedance

In AC circuits at frequency $f$, impedances are complex numbers:
- Resistor: $Z_R = R$ (real)
- Inductor: $Z_L = j\omega L$ (purely imaginary, positive)
- Capacitor: $Z_C = 1/(j\omega C) = -j/(\omega C)$ (purely imaginary, negative)

```python
# --- Series RLC Impedance ---
R  = 100     # ohms
L  = 0.01    # henries (10 mH)
C  = 1e-6    # farads (1 µF)
f  = 1000    # Hz
omega = 2 * np.pi * f

Z_R = R + 0j
Z_L = 1j * omega * L
Z_C = 1 / (1j * omega * C)

Z_total = Z_R + Z_L + Z_C

print(f'At f = {f} Hz  (ω = {omega:.1f} rad/s):')
print(f'  Z_R = {Z_R}')
print(f'  Z_L = {Z_L:.2f}')
print(f'  Z_C = {Z_C:.2f}')
print(f'  Z_total = {Z_total:.2f}')
print(f'  |Z_total| = {abs(Z_total):.2f} Ω   (this is what limits current)')
print(f'  Phase = {np.degrees(cmath.phase(Z_total)):.2f}° (voltage leads/lags current)')

# --- Resonant frequency: when Im(Z) = 0 → XL = XC ---
f_resonant = 1 / (2 * np.pi * np.sqrt(L * C))
print(f'\nResonant frequency: {f_resonant:.1f} Hz')
print('  At resonance, Z = R + 0j → purely resistive, minimum impedance')

# --- Sweep frequency and plot impedance ---
freqs = np.logspace(2, 5, 500)   # 100 Hz to 100 kHz
omegas = 2 * np.pi * freqs
Z_sweep = R + 1j*omegas*L + 1/(1j*omegas*C)

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(8, 6))
ax1.semilogx(freqs, np.abs(Z_sweep), 'b-', linewidth=2)
ax1.axvline(f_resonant, color='r', linestyle='--', label=f'f₀ = {f_resonant:.0f} Hz')
ax1.set_ylabel('|Z| (Ω)'); ax1.set_title('RLC Series Impedance vs Frequency')
ax1.legend(); ax1.grid(True, alpha=0.3)

ax2.semilogx(freqs, np.degrees(np.angle(Z_sweep)), 'g-', linewidth=2)
ax2.axvline(f_resonant, color='r', linestyle='--')
ax2.axhline(0, color='gray', linewidth=0.8)
ax2.set_xlabel('Frequency (Hz)'); ax2.set_ylabel('Phase (°)')
ax2.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
