## 1. Writing and Reading Plain Text Files

Python's built-in `open()` function is the general-purpose tool for any text file. The standard, safe pattern is `with open(filename, mode) as f:` — this automatically closes the file when the block ends, even if an error happens partway through.

**File modes you'll use constantly:**
- `"w"` — write (creates the file if it doesn't exist; **overwrites** it completely if it does)
- `"r"` — read (the default if you don't specify a mode)
- `"a"` — append (adds to the end of an existing file, without erasing what's already there)


```python
with open("demo1.txt", "w") as f:
    f.write("Line one\n")
    f.write("Line two\n")

print("File written.")
```

Reading the whole file at once with `.read()`:

```python
with open("demo1.txt") as f:
    content = f.read()

print(repr(content))   # repr() shows the \n characters explicitly, useful for debugging
```

## 2. Reading Line by Line

Reading the whole file at once is fine for small files, but looping over the file object directly (or using `.readlines()`) is often more useful when you need to process each line separately.

```python
with open("demo1.txt") as f:
    for line in f:
        print(line.strip())   # .strip() removes the trailing newline character
```

`.readlines()` gives you the same lines back as an actual list, if you need to store them rather than just loop through once:

```python
with open("demo1.txt") as f:
    lines = f.readlines()

print(lines)
print("Number of lines:", len(lines))
```

## 3. Appending to a File

Opening a file with `"w"` **erases** its previous contents — a very common source of accidentally-lost data. If you want to add to a file without erasing it, use `"a"` (append) instead.

```python
with open("demo1.txt", "a") as f:
    f.write("Line three\n")

with open("demo1.txt") as f:
    print(f.read())
```

**Careful:** running a `"w"`-mode cell more than once will silently overwrite your file every time, with no warning. This is one of the most common file-handling bugs — always double check which mode you actually want before running.

## 4. Loading and Saving Numeric Data with NumPy

For a file that's *purely* numbers (no text, no headers), `np.loadtxt()` and `np.savetxt()` are the simplest tools — they go directly to and from a NumPy array.

```python
arr = np.array([1.5, 2.5, 3.5, 4.5])

np.savetxt("demo2.txt", arr, fmt="%.1f")

with open("demo2.txt") as f:
    print(f.read())
```

```python
arr2 = np.loadtxt("demo2.txt")
print(arr2)
print(type(arr2))
```

## 5. CSV Files with the `csv` Module

When your data mixes text and numbers, or needs column headers, the built-in `csv` module gives you row-by-row control. Always use `newline=""` when opening a file for `csv.writer` — this avoids extra blank lines appearing on some systems.

```python
import csv

with open("demo3.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["name", "score"])   # header row
    writer.writerow(["Alice", 95])
    writer.writerow(["Bob", 88])

print("CSV written.")
```

```python
with open("demo3.csv") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
```

Note: every value read back from a CSV file is a **string** — including the numbers. If you need to do math with them, cast explicitly (e.g., `int(row[1])`), the same casting rule from Week 1.

## 6. CSV Files with pandas

When you want labeled columns and plan to analyze the data further (filter, sort, compute statistics), `pandas` is usually the most convenient tool.

```python
import pandas as pd

df = pd.read_csv("demo3.csv")
print(df)
print()
print("Average score:", df["score"].mean())
```
