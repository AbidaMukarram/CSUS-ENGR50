## 1. Python as a Calculator

Before Python does anything fancy, it's just a very capable calculator. You can type an arithmetic expression directly into a code cell and run it.

| Operator | Meaning | Example |
|---|---|---|
| `+` | addition | `2 + 3` |
| `-` | subtraction | `5 - 1` |
| `*` | multiplication | `4 * 3` |
| `/` | division (always returns a float) | `7 / 2` |
| `//` | floor division (rounds down to an integer) | `7 // 2` |
| `%` | modulo (remainder after division) | `7 % 2` |
| `**` | exponent (power) | `2 ** 3` |

Python follows the standard order of operations (PEMDAS) — parentheses first, then exponents, then multiplication/division, then addition/subtraction.

````python
2 + 3
````

Notice that Jupyter automatically displays the result of the **last line** in a cell, even without `print()`. This only works for the last line — if you want to see multiple results, you need `print()`.

````python
print(7 / 2)    # true division -- always a float
print(7 // 2)   # floor division -- rounds down
print(7 % 2)    # modulo -- the remainder
print(2 ** 3)   # exponent -- 2 to the power of 3
````

````python
# Order of operations matters!
print(2 + 3 * 4)      # multiplication happens first: 2 + 12 = 14
print((2 + 3) * 4)    # parentheses override that: 5 * 4 = 20
````

## 2. Variables

A **variable** is a name that stores a value so you can reuse it later, instead of retyping the value every time. In Python, you create a variable with a single equals sign `=` (this is *assignment*, not mathematical equality!).

**Naming rules:**
- Must start with a letter or underscore (not a number)
- Can contain letters, numbers, and underscores
- Case-sensitive (`age` and `Age` are different variables)
- Cannot be a Python reserved word (like `for`, `if`, `print`)

**Naming convention:** use `snake_case` (lowercase with underscores) for variable names in Python, e.g. `student_count`, not `StudentCount` or `studentCount`.


````python
collector_area = 2.4       # square meters
efficiency = 0.55
irradiance = 700           # watts per square meter

power_absorbed = collector_area * efficiency * irradiance
print(power_absorbed)
````

A variable can be **reassigned** at any time — Python doesn't lock in the original value or type:

````python
x = 5
print(x)
x = x + 1      # take the current value of x, add 1, store it back in x
print(x)
x = "now I'm a string"   # variables can even change type
print(x)
````

## 3. Data Types

Every value in Python has a **type**. The four most common basic types you'll use constantly:

| Type | Name | Example | Notes |
|---|---|---|---|
| `int` | integer | `42`, `-7` | whole numbers, no decimal point |
| `float` | floating-point number | `3.14`, `-0.5`, `2.0` | has a decimal point |
| `str` | string | `"hello"`, `'x'` | text, in single or double quotes |
| `bool` | boolean | `True`, `False` | exactly one of two values (capitalized!) |

You can check any value's type with the built-in `type()` function.


````python
print(type(42))
print(type(3.14))
print(type("hello"))
print(type(True))
````

A common surprise: `5` and `5.0` are *not* the same type, even though they represent the same number:

````python
print(type(5))
print(type(5.0))
print(5 == 5.0)   # they ARE considered equal in value, just not the same type
````

## 4. Type Casting

**Casting** means explicitly converting a value from one type to another, using the type name as a function: `int(...)`, `float(...)`, `str(...)`, `bool(...)`.

This comes up constantly when working with real data — e.g., a value read from a file or user input often arrives as a `str`, even if it "looks like" a number, and you have to cast it before doing math with it.


````python
age_str = "20"
print(type(age_str))

age_int = int(age_str)    # cast string -> int
print(type(age_int))
print(age_int + 5)         # now math works
````

Casting `float` &rarr; `int` **truncates** (chops off the decimal part) rather than rounding:

````python
print(int(9.99))   # truncates to 9, NOT rounded to 10
print(int(9.01))   # also 9
print(round(9.99)) # round() is the function to use if you actually want rounding
````

Not every cast is valid — casting a non-numeric string to `int` raises an error:

````python
try:
    int("hello")
except ValueError as e:
    print("This failed, as expected:", e)
````

## 5. Formatted Printing (Python's version of `printf`)

If you've used C, you might know `printf("%d apples", count)`. Python has an equivalent — and a more modern, easier-to-read alternative.

### Old style: `%` formatting (works like C's `printf`)


````python
name = "Alex"
score = 92.5
print("Student %s scored %.1f points" % (name, score))
# %s = string, %.1f = float rounded to 1 decimal place, %d = integer
````

### Modern style: f-strings (recommended)

An **f-string** is a string with an `f` right before the opening quote. Anything inside `{}` is evaluated as Python code and inserted directly into the string. This is the standard, recommended way to format strings in modern Python (3.6+), and what you'll see used throughout this course.


````python
name = "Alex"
score = 92.5
print(f"Student {name} scored {score} points")
````

f-strings also support format specifiers after a colon `:`, for controlling decimal places, padding, etc. — this is the direct equivalent of `printf`'s `%.2f`-style formatting:

````python
pi = 3.14159265
print(f"Pi rounded to 2 decimal places: {pi:.2f}")
print(f"Pi rounded to 4 decimal places: {pi:.4f}")

count = 7
print(f"Count padded to 3 digits: {count:03}")   # 007

large_num = 1234567
print(f"With thousands separator: {large_num:,}")
````

## 6. Complex Number Arithmetic

Python has built-in support for complex numbers — no import needed for basic arithmetic. A complex number is written with a `j` (not `i`, as in math) directly after the imaginary part's coefficient, e.g. `3 + 4j`.

| Operation | Syntax | Notes |
|---|---|---|
| Create | `3 + 4j` or `complex(3, 4)` | both create the same value |
| Real part | `z.real` | always returns a `float` |
| Imaginary part | `z.imag` | always returns a `float` |
| Conjugate | `z.conjugate()` | flips the sign of the imaginary part |
| Magnitude | `abs(z)` | $\sqrt{\text{real}^2 + \text{imag}^2}$ |
| Add, subtract, multiply, divide | `+`, `-`, `*`, `/` | work exactly like normal arithmetic |


````python
z1 = 3 + 4j
z2 = 1 - 2j

print("z1 =", z1)
print("z2 =", z2)
print("type(z1):", type(z1))
````

````python
print("z1 + z2 =", z1 + z2)
print("z1 - z2 =", z1 - z2)
print("z1 * z2 =", z1 * z2)
print("z1 / z2 =", z1 / z2)
````

````python
print("Real part of z1:", z1.real)
print("Imaginary part of z1:", z1.imag)
print("Conjugate of z1:", z1.conjugate())
print("Magnitude of z1 (|z1|):", abs(z1))   # sqrt(3^2 + 4^2) = 5.0
````

A common gotcha: writing `3 + 4i` (using `i` like in math class, with no `*`) is actually a **SyntaxError** in Python — it must be `j`, written directly against the number with no space:

````python
try:
    exec("bad = 3 + 4i")
except SyntaxError as e:
    print("This failed, as expected:", e)
````

## 7. Floating-Point Numbers, the `math` Library, and Why `==` Can Lie to You

You've already seen that `float` is one of Python's basic types. This section goes deeper: floats aren't stored with perfect precision, which causes some genuinely surprising behavior you need to know about before you write code that compares numbers.

### More Floating-Point Arithmetic Examples

Floats support the same operators as integers, but division, roots, and repeating decimals often produce long, imprecise-looking results:

````python
print(1 / 3)
print(2 / 3)
print(10 / 4)
print(2 ** 0.5)      # square root of 2, using the exponent operator
print(3.3 - 3.1)     # you might expect exactly 0.2 -- see what you actually get
````

That last one is the first hint of the problem: `3.3 - 3.1` doesn't print as a clean `0.2`. Computers store floats in binary, and most decimal fractions (like 0.1, 0.2, 0.3) **cannot be represented exactly** in binary — the same way $\frac{1}{3}$ can't be written exactly as a finite decimal. Python is doing the math correctly; it's just displaying the tiny leftover imprecision.

### The Classic Floating-Point Comparison Trap

This is one of the most common bugs in scientific and engineering code: comparing two floats with `==` when you actually meant "close enough."

````python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)
````

That printed `False`! Let's see why, by looking at the *actual* stored value with more decimal places using `repr()`:

````python
print(repr(0.1 + 0.2))
print(repr(0.3))
````

`0.1 + 0.2` is actually stored as `0.30000000000000004` — off from `0.3` by about $4 \times 10^{-17}$. That's a tiny error, but `==` checks for *exact* equality, so even a difference this small makes the comparison `False`. This is not a Python bug — it happens in essentially every programming language, because it's a consequence of binary floating-point representation (IEEE 754), not a language design choice.

**A second classic example**, familiar from earlier in the course: adding a small number to a very large one can lose it entirely.

````python
a = 1e16
print((a + 1) - a)   # you'd expect 1.0 -- but the '+1' doesn't fit in the available precision
````

### The Fix: `math.isclose()` Instead of `==`

The correct way to compare floats is to check whether they're *close enough*, not exactly equal. Python's built-in `math` library provides `math.isclose()` for exactly this purpose.

````python
import math

print(math.isclose(0.1 + 0.2, 0.3))              # True -- this is the fix
print(math.isclose(1.0000001, 1.0000002))          # False by default -- very tight tolerance
print(math.isclose(1.0000001, 1.0000002, abs_tol=1e-5))  # True -- loosened tolerance
````

`math.isclose()` has two tolerance settings you can control: `rel_tol` (relative tolerance, a percentage of the larger number) and `abs_tol` (absolute tolerance, a fixed cushion — important when comparing numbers near zero, where a purely relative tolerance breaks down). For most engineering calculations, setting an explicit `abs_tol` appropriate to your problem is the safest habit.

### The `math` Library: Common Functions

Python's built-in `math` library provides functions and constants beyond basic arithmetic. You must `import math` before using it.

````python
import math

print(math.sqrt(2))          # square root
print(math.pi)                # the constant pi
print(math.e)                 # Euler's number
print(math.floor(4.7))        # rounds down to the nearest integer
print(math.ceil(4.2))         # rounds up to the nearest integer
print(math.factorial(5))      # 5! = 5*4*3*2*1
````

Note: `math.sqrt()` only works on real, non-negative numbers — for complex numbers (Section 6 above), you'd use the separate `cmath` library instead, or simply use `**0.5`, which works for both.

````python
try:
    math.sqrt(-4)
except ValueError as e:
    print("This failed, as expected:", e)

print((-4) ** 0.5)   # this works and returns a complex result instead
````

## 9. Simple String "Math": `+` and `*`

Strings support two of the same operators you've been using on numbers — but they do something different: `+` **concatenates** (joins) strings together, and `*` **repeats** a string a given number of times.

*(This is just a preview — we'll explore strings in much more detail, including indexing, slicing, and string methods, in a later notebook.)*

````python
# + concatenates strings
first = "Py"
second = "thon"
print(first + second)

greeting = "Hello" + ", " + "world" + "!"
print(greeting)
````

````python
# * repeats a string
print("ab" * 3)
print("=" * 20)      # a quick way to print a divider line
print("Ha" * 4 + "!")
````

Note: `+` requires **both sides to be strings** — you can't directly add a string and a number, even if the number is inside a variable. This is a very common early error:

````python
try:
    print("Age: " + 21)
except TypeError as e:
    print("This failed, as expected:", e)

# The fix: cast the number to a string first
print("Age: " + str(21))
````
