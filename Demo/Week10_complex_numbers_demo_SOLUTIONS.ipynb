### 🎯 Your Turn 1 — GPS Coordinates as Complex Numbers

Engineers sometimes represent a 2D position as a complex number:  
**x-coordinate = real part**, **y-coordinate = imaginary part**.

Three campus buildings are at these (x, y) positions (in meters from the main gate):
- **Engineering Hall**: (120, 85)
- **Library**: (−40, 200)
- **Cafeteria**: (95, −30)

**Tasks:**
1. Create each building as a complex number (e.g. `eng_hall = 120 + 85j`)
2. Print the real and imaginary parts of each to confirm
3. A student walks from Engineering Hall to the Library.  
   Compute the **displacement vector** `d = library - eng_hall`
4. The displacement's magnitude `abs(d)` is the straight-line walking distance. Print it.
5. How far is the Cafeteria from the Library? Compute and print it.

**Think about it:** Why is subtraction of complex numbers the same as finding a displacement vector?


```python
# ══════════════════════════════════════════════════════════════════════════════
# YOUR TURN 1 — GPS Campus Navigation
# ══════════════════════════════════════════════════════════════════════════════
#
# WHY COMPLEX NUMBERS INSTEAD OF JUST TWO REAL VARIABLES?
# ─────────────────────────────────────────────────────────
# You could store every location as (x, y) with two separate floats.
# That works, but then distance is:
#   dist = math.sqrt((x2-x1)**2 + (y2-y1)**2)   ← messy!
#
# With complex numbers, the same thing is just:
#   dist = abs(z2 - z1)                           ← beautiful!
#
# The complex number z = x + yj is essentially a 2-D vector.
# The real part  → East-West  (x-axis)
# The imaginary part → North-South (y-axis)
#
# Operations you get for FREE by treating positions as complex numbers:
#   • Subtraction  →  displacement vector  (direction AND distance in one step)
#   • abs(z)       →  Euclidean distance (straight-line)
#   • z1 + z2      →  vector addition (great for navigation legs)
#   • cmath.phase  →  compass bearing
#
# A "real" number only has ONE component, so it can describe a 1-D position
# (like position along a single road).  Campus navigation is 2-D — you need
# both East/West AND North/South, so you need a 2-D number type. Complex
# numbers are the most natural one Python offers.

# ── Step 1: Create the buildings as complex numbers ───────────────────────
# Convention: (0, 0) = main campus gate
#   Real part  = meters East  of gate  (negative = West)
#   Imag part  = meters North of gate  (negative = South)
eng_hall  = 120 + 85j   # 120 m East, 85 m North
library   = -40 + 200j  # 40 m West, 200 m North
cafeteria =  95 - 30j   # 95 m East, 30 m South

for name, pos in [("Eng Hall",  eng_hall),
                  ("Library",   library),
                  ("Cafeteria", cafeteria)]:
    print(f"{name:12s}  East = {pos.real:>7.1f} m   North = {pos.imag:>7.1f} m")

# ── Step 2: Displacement vector from Eng Hall to Library ─────────────────
#
# WHAT IS A DISPLACEMENT VECTOR?
# ────────────────────────────────
# A displacement vector answers: "If I'm standing at point A, how far and
# in what direction must I walk to reach point B?"
#
# In 2-D:  displacement = B - A = (B_x - A_x) + (B_y - A_y)j
#
# Example:  Library - Eng Hall
#         = (-40 + 200j) - (120 + 85j)
#         = (-40 - 120) + (200 - 85)j
#         = -160 + 115j
#
# Reading that answer:  go 160 m WEST (negative East) and 115 m NORTH.
# It tells you the direction AND the components of the journey.
# 'abs()' of this vector gives the straight-line walking distance.

d = library - eng_hall           # displacement vector  (-160 + 115j)
dist_to_library = abs(d)         # Euclidean distance = sqrt(160² + 115²)
print(f"\nDisplacement Eng Hall → Library: {d}")
print(f"  → go {abs(d.real):.1f} m {'West' if d.real < 0 else 'East'}, "
      f"{abs(d.imag):.1f} m {'North' if d.imag > 0 else 'South'}")
print(f"  → straight-line distance = {dist_to_library:.1f} m")

# WHY NOT JUST USE abs(library - eng_hall) DIRECTLY?
# Because 'd' also carries directional info (the angle), useful if you need
# a compass bearing: cmath.phase(d) in degrees.
print(f"  → compass bearing = {math.degrees(cmath.phase(d)):.1f}°  (measured CCW from East)")

# ── Step 3: Cafeteria to Library distance ─────────────────────────────────
dist_caf_lib = abs(library - cafeteria)
print(f"\nCafeteria → Library straight-line: {dist_caf_lib:.1f} m")

# ── Why does subtraction give distance? ───────────────────────────────────
# abs(z) = sqrt(real² + imag²)   ← Pythagorean theorem!
# abs(B - A) = sqrt((B.real-A.real)² + (B.imag-A.imag)²)
# That IS the distance formula from geometry — complex subtraction + abs()
# encodes the Pythagorean theorem in one elegant expression.
```

## 2. Arithmetic Operations

```python
z1 = 3 + 4j
z2 = 1 - 2j

# Addition
print(f'z1 + z2 = {z1 + z2}')    # (3+1) + (4-2)j = 4 + 2j

# Subtraction
print(f'z1 - z2 = {z1 - z2}')    # (3-1) + (4+2)j = 2 + 6j

# Multiplication  (FOIL: (3+4j)(1-2j))
product = z1 * z2
print(f'z1 * z2 = {product}')     # 3-6j+4j-8j² = 3-2j+8 = 11 + 2j

# Division
quotient = z1 / z2
print(f'z1 / z2 = {quotient:.4f}')  # multiply by conjugate of denominator

# Powers
print(f'j^2 = {1j**2}')           # should be -1
print(f'j^3 = {1j**3}')           # should be -j
print(f'j^4 = {1j**4}')           # should be 1
```

### 🎯 Your Turn 2 — Drone Navigation with Phasors

A delivery drone flies a 3-leg journey. Each leg is described as a **complex displacement**  
(real = East/West, imaginary = North/South, in meters):

- **Leg 1:** Fly `200 + 150j` m (northeast toward the warehouse)
- **Leg 2:** Fly `−80 + 220j` m (northwest to avoid a no-fly zone)
- **Leg 3:** Fly `50 − 100j` m (southeast to the delivery point)

**Tasks:**
1. Create the three legs as complex numbers
2. Compute the **total displacement** = sum of all three legs
3. Print the final position (how far East and how far North of the start)
4. Compute the **total path length** (sum of magnitudes of each leg — actual distance flown)
5. Compute the **straight-line distance** from start to finish (`abs(total displacement)`)
6. The drone wastes energy flying extra. Compute:  
   `efficiency = straight_line_distance / total_path_length`  
   Print it as a percentage. A value close to 100% means very direct routing.

**Think about it:** Why is the straight-line distance always ≤ total path length?  
(This is the triangle inequality for complex numbers!)


```python
# ══════════════════════════════════════════════════════════════════════════════
# YOUR TURN 2 — Drone Delivery Navigation
# ══════════════════════════════════════════════════════════════════════════════
#
# WHY COMPLEX NUMBERS FOR DRONE PATHS?
# ──────────────────────────────────────
# A drone moves in 2-D (we're ignoring altitude here).  Each flight LEG is
# a 2-D vector: it has a direction and a distance.
#
# If you stored legs as pairs of real numbers, you'd need:
#   leg1_x, leg1_y = 200, 150
#   total_x = leg1_x + leg2_x + leg3_x   ← two separate additions
#   total_y = leg1_y + leg2_y + leg3_y
#   straight_line = math.sqrt(total_x**2 + total_y**2)
#
# With complex numbers, vector addition is just the + operator:
#   total = leg1 + leg2 + leg3            ← one line, same result
#   straight_line = abs(total)
#
# This scales beautifully to dozens of legs with sum([all_legs]).
# Real numbers can't add 2-D vectors — they only have one axis.

import math, cmath

# ── Step 1: Define each flight leg as a complex displacement ──────────────
# Format: (meters East) + (meters North)j
# Negative real = flying West; negative imag = flying South
leg1 = 200 + 150j   # fly 200 m East and 150 m North
leg2 = -80 + 220j   # fly 80 m West and 220 m North
leg3 =  50 - 100j   # fly 50 m East and 100 m South

print("Individual legs:")
for i, leg in enumerate([leg1, leg2, leg3], 1):
    direction = cmath.phase(leg)
    print(f"  Leg {i}: {leg}   |distance| = {abs(leg):.1f} m   bearing = {math.degrees(direction):.1f}°")

# ── Step 2: Net displacement (vector sum) ─────────────────────────────────
# Adding complex numbers adds the real parts together AND the imaginary parts
# together — exactly like adding 2-D vectors component by component.
total = leg1 + leg2 + leg3
print(f"\nNet displacement: {total}")
print(f"  Final position: {total.real:.1f} m East, {total.imag:.1f} m North of start")

# ── Step 3: Total distance actually flown ────────────────────────────────
# We need abs() of EACH leg separately, then add them.
# (abs of the sum ≠ sum of the abs values — see Triangle Inequality below)
path_length = abs(leg1) + abs(leg2) + abs(leg3)
print(f"\nTotal path flown: {path_length:.1f} m")

# ── Step 4: Straight-line distance start → finish ─────────────────────────
straight = abs(total)
print(f"Straight-line distance: {straight:.1f} m")

# ── Step 5: Routing efficiency ────────────────────────────────────────────
# If the drone could fly straight from start to finish it would cover
# 'straight' meters.  It actually flew 'path_length' meters.
# Efficiency < 100% whenever the route is not a straight line.
efficiency = (straight / path_length) * 100
print(f"Routing efficiency: {efficiency:.1f}%")
print(f"  ({100-efficiency:.1f}% extra distance due to the zigzag route)")

# ── THE TRIANGLE INEQUALITY (key insight) ────────────────────────────────
# |leg1 + leg2 + leg3|  ≤  |leg1| + |leg2| + |leg3|
# abs(total)            ≤  path_length
#
# The straight-line distance can NEVER exceed the total path flown.
# Equality holds only when all legs point in the exact same direction.
# This is why routing efficiency is always between 0% and 100%.
print(f"\nTriangle Inequality check: {straight:.2f} ≤ {path_length:.2f}  → {straight <= path_length}")
```

## 3. Conjugate and Magnitude

```python
z = 3 + 4j

# Conjugate: flip sign of imaginary part
z_conj = z.conjugate()
print(f'z        = {z}')
print(f'conj(z)  = {z_conj}')

# Magnitude (absolute value): |z| = sqrt(a² + b²)
magnitude = abs(z)              # Python built-in
magnitude_manual = (z.real**2 + z.imag**2) ** 0.5
print(f'|z| = {magnitude}')    # sqrt(9 + 16) = sqrt(25) = 5

# Verify: z * conj(z) = |z|²
print(f'z * conj(z) = {z * z_conj}')   # should be 25 + 0j

# Phase angle (argument)
phase = cmath.phase(z)
print(f'Phase = {phase:.4f} rad = {np.degrees(phase):.2f}°')  # atan(4/3) ≈ 53.13°
```

### 🎯 Your Turn 3 — Noise Cancellation in Headphones

Active noise-canceling headphones work by detecting an incoming sound wave and  
playing back its **exact negative** (the conjugate-like "anti-noise").

For a pure tone, the sound wave is represented as:  
`noise = amplitude + phase_shift·j`

The anti-noise must have the **same magnitude** but **opposite sign** (flip both parts):  
`anti_noise = -noise`  
When added together, they cancel: `noise + anti_noise = 0 + 0j` ✓

**Tasks:**
1. A microphone detects `noise = 3.2 + 1.8j` (arbitrary units). Create it.
2. Compute `anti_noise = -noise`. Print both and verify `noise + anti_noise == 0j`.
3. The microphone isn't perfect — it adds a small error: `measured = 3.1 + 2.0j`.  
   Compute the **residual** = `noise + (-measured)`. What is `abs(residual)`?  
   This is how much noise still gets through — the cancellation error.
4. The headphone specs say cancellation is effective when `abs(residual) < 0.5`.  
   Print whether this headphone **passes** or **fails** the spec.
5. What percentage of the original noise magnitude gets through?  
   `leakage_pct = abs(residual) / abs(noise) * 100`

**Think about it:** Why do we care about `abs(residual)` rather than just `residual` itself?


```python
# ══════════════════════════════════════════════════════════════════════════════
# YOUR TURN 3 — Active Noise Cancellation (Headphones)
# ══════════════════════════════════════════════════════════════════════════════
#
# WHY DO SOUND SIGNALS NEED COMPLEX NUMBERS?
# ────────────────────────────────────────────
# Sound is a wave.  Every wave has TWO independent properties:
#   1. AMPLITUDE  — how loud it is (the height of the wave)
#   2. PHASE      — where in its cycle it currently sits (timing offset)
#
# A real number can store amplitude, but it cannot store phase.
# If you only track amplitude you don't know WHEN the wave peaks.
#
# Complex representation:  signal = amplitude * e^(j*phase)
#   • abs(signal)          → amplitude  (how loud)
#   • cmath.phase(signal)  → phase angle (timing)
#
# For noise cancellation, you need both.  The anti-noise must have:
#   • the SAME amplitude as the noise   (to cancel it out completely)
#   • OPPOSITE phase (180° shift)       (so peaks align with troughs)
#
# In complex arithmetic, "180° phase shift" is simply multiplication by -1:
#   anti_noise = -noise
# That single negation simultaneously flips the phase AND preserves the amplitude.
# With real numbers you'd need separate trigonometric calculations for phase.

import math, cmath

# ── Step 1: The incoming noise signal ─────────────────────────────────────
# The complex number encodes BOTH amplitude and phase in one value.
noise = 3.2 + 1.8j
print(f"Noise signal:     {noise}")
print(f"  Amplitude (loudness):  {abs(noise):.4f}")
print(f"  Phase (timing offset): {math.degrees(cmath.phase(noise)):.2f}°")

# ── Step 2: Perfect anti-noise = exact negation ───────────────────────────
# Negating a complex number flips both real and imaginary parts,
# which is equivalent to shifting the phase by exactly 180°.
anti_noise = -noise
print(f"\nAnti-noise signal: {anti_noise}")
print(f"  Phase of anti-noise:   {math.degrees(cmath.phase(anti_noise)):.2f}°  (exactly 180° shifted)")
print(f"  noise + anti_noise   = {noise + anti_noise}   ← perfect silence!")
print(f"  Amplitude preserved:   {abs(anti_noise):.4f}  (same loudness, opposite phase)")

# ── Step 3: Real-world imperfection ───────────────────────────────────────
# The microphone that measures the noise has a tiny error.
# The 'measured' signal is close to 'noise' but not perfect.
measured  = 3.1 + 2.0j
residual  = noise + (-measured)       # what's left after cancellation attempt
print(f"\nMeasured noise:   {measured}")
print(f"Residual noise:   {residual}  ← not exactly 0 because measurement was imperfect")
print(f"Residual amplitude: {abs(residual):.4f}")

# ── Step 4: Does it pass the specification? ────────────────────────────────
# We care about |residual|, not residual itself.
# A residual of 0.1+0.2j and -0.2-0.1j are equally "loud" — both have
# the same energy regardless of phase angle.
# The spec says: total residual noise energy must be below 0.5 units.
spec_limit = 0.5
passes = abs(residual) < spec_limit
print(f"\nSpec limit: {spec_limit}")
print(f"Passes spec: {passes}   (residual amplitude = {abs(residual):.4f})")

# ── Step 5: Leakage percentage ─────────────────────────────────────────────
# How much of the original noise "leaked through" after cancellation?
leakage_pct = abs(residual) / abs(noise) * 100
print(f"Noise leakage: {leakage_pct:.1f}%  of original noise amplitude remains")
print(f"Noise reduction: {100-leakage_pct:.1f}%")

# ── Summary: Why not real numbers? ────────────────────────────────────────
# Real number approach would need:
#   residual_amp   = sqrt((noise_amp*cos(noise_phase) - measured_amp*cos(measured_phase))**2 +
#                         (noise_amp*sin(noise_phase) - measured_amp*sin(measured_phase))**2)
# Complex number approach:
#   residual_amp   = abs(noise - measured)   ← identical result, far simpler!
```

## 4. Polar Form

Any complex number can be written as $z = r \cdot e^{j\theta} = r(\cos\theta + j\sin\theta)$

```python
z = 3 + 4j

# --- Rectangular → Polar ---
r, theta = cmath.polar(z)   # returns (magnitude, phase_in_radians)
print(f'Rectangular: {z}')
print(f'Polar:  r = {r}, θ = {theta:.4f} rad ({np.degrees(theta):.2f}°)')

# --- Polar → Rectangular ---
z_back = cmath.rect(r, theta)   # converts (r, theta) → complex
print(f'Back to rectangular: {z_back:.4f}')

# --- Euler's formula: e^(jθ) ---
theta = np.pi / 4   # 45 degrees
z_euler = cmath.exp(1j * theta)
print(f'\ne^(j·π/4) = {z_euler:.4f}')
print(f'Which is:   cos(45°) + j·sin(45°) = {np.cos(theta):.4f} + {np.sin(theta):.4f}j')

# --- Visualize on the complex plane ---
fig, ax = plt.subplots(figsize=(6, 6))
ax.axhline(0, color='gray', linewidth=0.8)
ax.axvline(0, color='gray', linewidth=0.8)
ax.plot(z.real, z.imag, 'bo', markersize=10, label='z = 3+4j')
ax.annotate('z = 3+4j', xy=(z.real, z.imag), xytext=(z.real+0.3, z.imag+0.2), fontsize=12)
ax.arrow(0, 0, z.real, z.imag, head_width=0.15, head_length=0.15, fc='blue', ec='blue')
ax.set_xlim(-1, 6); ax.set_ylim(-1, 6)
ax.set_xlabel('Real'); ax.set_ylabel('Imaginary')
ax.set_title('Complex Number on the Complex Plane')
ax.grid(True, alpha=0.3)
ax.set_aspect('equal')
plt.tight_layout()
plt.show()
print(f'Magnitude |z| = {r:.1f}, Angle θ = {np.degrees(theta):.1f}°')
```

### 🎯 Your Turn 4 — Robot Arm Rotation

A robot arm has its tip at position `tip = 5 + 0j` meters (pointing East).

**Key insight:** Multiplying a complex number by $e^{j\theta}$ **rotates it by θ radians**.  
This is because $e^{j\theta} = \cos\theta + j\sin\theta$.

**Tasks:**
1. Rotate the tip **45°** counterclockwise:  
   `tip_rotated = tip * cmath.exp(1j * math.radians(45))`  
   Print the new (x, y) position and verify `abs(tip_rotated) == abs(tip)` (length unchanged).
2. Rotate the original tip through **0°, 60°, 120°, 180°, 240°, 300°** (a full hexagon).  
   Print each position rounded to 2 decimal places.
3. Plot all 6 tip positions as dots connected by lines (the hexagon shape).  
   Also draw the robot arm as an arrow from origin to each point.
4. A second arm segment of length 3 m is attached at the end of the first.  
   At 45° rotation of the first arm, the second arm points at **30°** from horizontal.  
   Compute the **absolute position** of the second arm's tip:  
   `tip2 = tip_rotated + 3 * cmath.exp(1j * math.radians(30))`

**Think about it:** Why is rotation by multiplication so elegant compared to 2D rotation matrices?


```python
# ══════════════════════════════════════════════════════════════════════════════
# YOUR TURN 4 — Robot Arm Rotation
# ══════════════════════════════════════════════════════════════════════════════
#
# WHY COMPLEX NUMBERS FOR ROTATION?
# ───────────────────────────────────
# In 2-D, rotating a point by angle θ normally requires a matrix multiply:
#
#   [x']   [cos θ  -sin θ] [x]
#   [y'] = [sin θ   cos θ] [y]
#
# That's 4 multiplications and 2 additions.
#
# With complex numbers, the SAME rotation is just one multiplication:
#   z_rotated = z * e^(j*θ)       ← Euler's formula encodes the rotation matrix
#
# Why does this work?  Euler's formula:  e^(jθ) = cos θ + j sin θ
# When you multiply two complex numbers you multiply their magnitudes and
# ADD their angles:
#   |z * e^(jθ)| = |z| * 1 = |z|      (length unchanged)
#   angle(z * e^(jθ)) = angle(z) + θ  (angle increased by θ)
#
# That IS a rotation — same length, new angle.  Elegantly encoded.
# A real number has no angle concept, so rotation is impossible with reals.

import math, cmath, numpy as np, matplotlib.pyplot as plt

# The arm starts horizontal (pointing East), length 5 m
# As a complex number: real part = length, imaginary part = 0
tip = 5 + 0j
print(f"Original tip position: ({tip.real:.2f}, {tip.imag:.2f})")
print(f"  Length = {abs(tip):.2f} m,  Angle = {math.degrees(cmath.phase(tip)):.1f}°")

# ── Step 1: Rotate 45° counterclockwise ──────────────────────────────────
# e^(jθ) is the "rotation operator" in complex arithmetic.
# θ must be in RADIANS (use math.radians to convert from degrees).
tip_45 = tip * cmath.exp(1j * math.radians(45))
print(f"\nAfter 45° rotation:  ({tip_45.real:.4f}, {tip_45.imag:.4f})")
print(f"  Length preserved: {abs(tip_45):.6f} m  (should equal 5.0 exactly)")
print(f"  New angle:        {math.degrees(cmath.phase(tip_45)):.1f}°  (should equal 45°)")

# ── Step 2: Sweep through 6 positions (every 60° → hexagon) ──────────────
print("\nHexagonal sweep (every 60°):")
angles = [0, 60, 120, 180, 240, 300]
positions = []
for deg in angles:
    pos = tip * cmath.exp(1j * math.radians(deg))
    positions.append(pos)
    print(f"  {deg:>3}°  →  ({pos.real:>7.4f},  {pos.imag:>7.4f})")

# NOTICE: all positions have length exactly 5.0 (rotation preserves length)
# This is NOT a coincidence — multiplication by e^(jθ) is a PURE rotation:
# it only changes direction, never magnitude.

# ── Step 3: Plot the hexagon ──────────────────────────────────────────────
xs = [p.real for p in positions] + [positions[0].real]   # close the shape
ys = [p.imag for p in positions] + [positions[0].imag]

fig, ax = plt.subplots(figsize=(6, 6))
ax.plot(xs, ys, 'b-o', linewidth=2, markersize=8, label='Tip positions')
ax.plot(0, 0, 'ks', markersize=10, label='Shoulder joint')
for p, deg in zip(positions, angles):
    ax.annotate(f'{deg}°', xy=(p.real, p.imag),
                xytext=(p.real*1.15, p.imag*1.15), fontsize=9, ha='center')
ax.axhline(0, color='gray', lw=0.5)
ax.axvline(0, color='gray', lw=0.5)
ax.set_xlim(-7, 7); ax.set_ylim(-7, 7)
ax.set_aspect('equal'); ax.grid(True, alpha=0.3)
ax.set_title('Robot Arm Tip — Hexagonal Path\n(Each position is a rotation of 5+0j by e^(jθ))')
ax.legend(); plt.tight_layout(); plt.show()

# ── Step 4: Two-segment arm ───────────────────────────────────────────────
# The first segment is already at 45°.  The second segment (length 3 m)
# is angled at 30° — but 30° from WHAT?  From the global x-axis.
# We just multiply: 3 * e^(j*30°) gives a vector in that direction.
# Then ADD to tip_45 to get the final tip position.
tip2 = tip_45 + 3 * cmath.exp(1j * math.radians(30))
print(f"\nTwo-segment arm:")
print(f"  Segment 1 tip (at 45°): ({tip_45.real:.4f}, {tip_45.imag:.4f})")
print(f"  Segment 2 added (at 30°): 3 * e^(j*30°) = ({3*math.cos(math.radians(30)):.4f}, {3*math.sin(math.radians(30)):.4f})")
print(f"  Final tip position: ({tip2.real:.4f}, {tip2.imag:.4f})")

# REAL-WORLD APPLICATION:
# This is exactly how robotics software (ROS, industrial controllers) computes
# forward kinematics for a chain of arm segments.  Each joint rotation is a
# complex multiply.  Chaining n rotations = multiplying n complex exponentials:
#   e^(jθ₁) * e^(jθ₂) = e^(j(θ₁+θ₂))   ← exponents just add!
# This is simpler than multiplying a chain of 2×2 rotation matrices.
```

## 5. Solving Polynomial Equations

Polynomials with real coefficients may have complex roots. `numpy.roots()` finds them all.

```python
# --- Example 1: x² + 1 = 0 (roots are ±j) ---
coeffs1 = [1, 0, 1]   # 1·x² + 0·x + 1
roots1 = np.roots(coeffs1)
print('x² + 1 = 0')
print(f'Roots: {roots1}')  # [0+1j, 0-1j]

# --- Example 2: x² + 2x + 5 = 0 (complex conjugate pair) ---
# Discriminant: b² - 4ac = 4 - 20 = -16  (negative → complex roots)
coeffs2 = [1, 2, 5]
roots2 = np.roots(coeffs2)
print('\nx² + 2x + 5 = 0')
for r in roots2:
    print(f'  Root: {r:.4f}  |magnitude: {abs(r):.4f}')

# --- Example 3: Characteristic equation of an RLC circuit ---
# LCs² + Rs + 1/C = 0 → s² + (R/L)s + 1/(LC) = 0
R, L, C = 100, 0.1, 1e-6   # ohms, henries, farads
coeffs3 = [1, R/L, 1/(L*C)]
s_poles = np.roots(coeffs3)
print(f'\nRLC circuit poles (s = σ ± jω):')
for s in s_poles:
    sigma = s.real
    omega = abs(s.imag)
    print(f'  s = {s:.1f}  →  σ={sigma:.1f} (decay rate),  ω={omega:.1f} rad/s (oscillation freq)')
```

### 🎯 Your Turn 5 — Where Does a Ball Land? (Complex Roots Edition)

A ball is thrown and its height is modeled by:
$$h(t) = -4.9t^2 + v_0 t + h_0$$

where $v_0$ is initial speed (m/s) and $h_0$ is starting height (m).

The ball hits the ground when $h(t) = 0$ — this is a quadratic we can solve with `np.roots()`.

**Tasks:**
1. For `v0 = 20 m/s`, `h0 = 2 m`, find the roots of `[-4.9, v0, h0]`.  
   One root is positive (when ball lands) and one is negative (ignore it — time can't be negative).  
   Print the landing time and how far it traveled horizontally if horizontal speed is 15 m/s.

2. Now try `v0 = 1 m/s`, `h0 = 0 m` (barely tossed from the ground).  
   Both roots should be real. What do they represent physically?

3. Try `v0 = 0 m/s`, `h0 = -5 m` (ball starts underground — physically impossible).  
   What kind of roots do you get (real or complex)?  
   Print whether each root is real using `abs(root.imag) < 1e-9`.

4. Find the **minimum** `v0` (starting from ground, `h0 = 0`) such that the ball stays in the air  
   for at least **3 seconds**. Try `v0 = 10, 12, 14, 15` and print the positive root for each.

**Think about it:** When does a physics problem have complex roots, and what does that mean physically?


```python
# ══════════════════════════════════════════════════════════════════════════════
# YOUR TURN 5 — Projectile Motion & Polynomial Roots
# ══════════════════════════════════════════════════════════════════════════════
#
# THE REAL-WORLD QUESTION
# ────────────────────────
# A ball thrown upward follows: h(t) = -4.9t² + v₀t + h₀
#   h  = height above ground (meters)
#   t  = time (seconds)
#   v₀ = initial upward speed
#   h₀ = starting height
#
# "When does it land?" means: when does h(t) = 0?
# → Solve:  -4.9t² + v₀t + h₀ = 0
# This is a QUADRATIC equation with roots that may be real or complex.
#
# WHY DO WE NEED COMPLEX NUMBERS FOR A REAL BALL?
# ─────────────────────────────────────────────────
# np.roots() always works in the complex number system (the "algebraic closure"
# of the reals). Most of the time the roots are real (the ball lands).
# But if the discriminant is negative (b² - 4ac < 0), the roots are COMPLEX —
# and that's physics telling you something: the scenario is IMPOSSIBLE.
# A ball that starts underground with no upward velocity can never reach h=0.
# Complex roots = "this event never happens in the real world."
#
# This is a profound insight: complex numbers don't just solve equations —
# they DIAGNOSE physically impossible situations.

import numpy as np

# ── Step 1: Normal launch (v₀=20 m/s, h₀=2 m above ground) ──────────────
v0, h0 = 20, 2
coeffs = [-4.9, v0, h0]    # polynomial coefficients [a, b, c] for at² + bt + c = 0
roots  = np.roots(coeffs)  # np.roots([a,b,c]) finds t where at²+bt+c = 0

print("── Launch: v₀=20 m/s, h₀=2 m ──")
print(f"Equation: -4.9t² + {v0}t + {h0} = 0")
print(f"Roots: {roots}")
print(f"Are roots real? {all(abs(r.imag) < 1e-9 for r in roots)}")

# Keep only the physically meaningful root (positive time = future event)
landing_time = max(r.real for r in roots if r.real > 0)
horiz_dist   = landing_time * 15    # horizontal speed = 15 m/s
print(f"Ball lands at t = {landing_time:.3f} s")
print(f"Horizontal distance = {horiz_dist:.2f} m")
print(f"(Negative root = {min(roots, key=lambda r: r.real).real:.3f} s is in the past — ignore it)")

# ── Step 2: Barely tossed from ground (v₀=1, h₀=0) ───────────────────────
print("\n── Barely tossed from ground: v₀=1, h₀=0 ──")
roots2 = np.roots([-4.9, 1, 0])    # h₀=0 means it starts at ground level
for r in roots2:
    is_real = abs(r.imag) < 1e-9
    meaning = "starts on ground (trivial)" if abs(r.real) < 1e-6 else "lands again"
    print(f"  root = {r:.6f}  ({meaning})")
# t=0 is a valid root: the ball IS on the ground at t=0 (before the toss).
# t≈0.204 is when it lands AFTER the tiny toss.
# Both roots are real because h₀=0 is not physically contradictory.

# ── Step 3: Starts underground — complex roots! ───────────────────────────
print("\n── Starts underground: v₀=0, h₀=-5 ──")
roots3 = np.roots([-4.9, 0, -5])   # h₀=-5: five meters underground, no initial velocity
print("Equation: -4.9t² + 0t + (-5) = 0  →  t² = -5/4.9  →  t = imaginary!")
for r in roots3:
    is_real = abs(r.imag) < 1e-9
    label = 'REAL' if is_real else '*** COMPLEX — no real solution! ***'
    print(f"  root = {r:.4f}  →  {label}")
print("\nInterpretation: The ball STARTS below ground and gravity pulls it")
print("further down. It NEVER reaches h=0. Complex roots = impossible scenario.")

# DISCRIMINANT ANALYSIS:
# b²-4ac = 0² - 4*(-4.9)*(-5) = -98  <  0  → complex roots, as expected
disc = 0**2 - 4*(-4.9)*(-5)
print(f"\nDiscriminant b²-4ac = {disc:.1f}  → {'real roots' if disc >= 0 else 'complex roots (impossible scenario)'}")

# ── Step 4: Engineering application — find minimum v₀ for 3 s airtime ────
print("\n── Engineering goal: find minimum v₀ for t_land ≥ 3 s (h₀=0) ──")
for v0_try in [10, 12, 14, 15]:
    r = np.roots([-4.9, v0_try, 0])
    t_land = max(x.real for x in r if x.real > 0)
    status = "✓ meets goal" if t_land >= 3 else "✗ too short"
    print(f"  v₀ = {v0_try} m/s  →  lands at {t_land:.3f} s  {status}")
```

## 6. Engineering Application: AC Circuit Impedance

In AC circuits at frequency $f$, impedances are complex numbers:
- Resistor: $Z_R = R$ (real)
- Inductor: $Z_L = j\omega L$ (purely imaginary, positive)
- Capacitor: $Z_C = 1/(j\omega C) = -j/(\omega C)$ (purely imaginary, negative)

```python
# --- Series RLC Impedance ---
R  = 100     # ohms
L  = 0.01    # henries (10 mH)
C  = 1e-6    # farads (1 µF)
f  = 1000    # Hz
omega = 2 * np.pi * f

Z_R = R + 0j
Z_L = 1j * omega * L
Z_C = 1 / (1j * omega * C)

Z_total = Z_R + Z_L + Z_C

print(f'At f = {f} Hz  (ω = {omega:.1f} rad/s):')
print(f'  Z_R = {Z_R}')
print(f'  Z_L = {Z_L:.2f}')
print(f'  Z_C = {Z_C:.2f}')
print(f'  Z_total = {Z_total:.2f}')
print(f'  |Z_total| = {abs(Z_total):.2f} Ω   (this is what limits current)')
print(f'  Phase = {np.degrees(cmath.phase(Z_total)):.2f}° (voltage leads/lags current)')

# --- Resonant frequency: when Im(Z) = 0 → XL = XC ---
f_resonant = 1 / (2 * np.pi * np.sqrt(L * C))
print(f'\nResonant frequency: {f_resonant:.1f} Hz')
print('  At resonance, Z = R + 0j → purely resistive, minimum impedance')

# --- Sweep frequency and plot impedance ---
freqs = np.logspace(2, 5, 500)   # 100 Hz to 100 kHz
omegas = 2 * np.pi * freqs
Z_sweep = R + 1j*omegas*L + 1/(1j*omegas*C)

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(8, 6))
ax1.semilogx(freqs, np.abs(Z_sweep), 'b-', linewidth=2)
ax1.axvline(f_resonant, color='r', linestyle='--', label=f'f₀ = {f_resonant:.0f} Hz')
ax1.set_ylabel('|Z| (Ω)'); ax1.set_title('RLC Series Impedance vs Frequency')
ax1.legend(); ax1.grid(True, alpha=0.3)

ax2.semilogx(freqs, np.degrees(np.angle(Z_sweep)), 'g-', linewidth=2)
ax2.axvline(f_resonant, color='r', linestyle='--')
ax2.axhline(0, color='gray', linewidth=0.8)
ax2.set_xlabel('Frequency (Hz)'); ax2.set_ylabel('Phase (°)')
ax2.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### 🎯 Your Turn 6 — Why Does Your Phone Charger Get Warm?

A USB phone charger has a simple internal circuit: a **resistor** R = 5 Ω in series  
with an **inductor** L = 2 mH. The wall outlet provides V = 120 V at 60 Hz.

**Tasks:**
1. Compute `Z_R = 5 + 0j` and `Z_L = 1j * omega * L` at 60 Hz.
2. Total impedance `Z_total = Z_R + Z_L`.  
   Print `|Z_total|` (how much it limits current) and the phase angle in degrees.
3. Current through the charger: `I = V / Z_total` where `V = 120 + 0j`.  
   Print `|I|` in amps.
4. Power dissipated **as heat** in the resistor only:  
   `P_heat = abs(I)**2 * R`  
   Print it in watts — this is why the charger gets warm!
5. Now compute with L = 0 (ideal resistor-only charger).  
   How does the heat change? Which dissipates **more** heat?
6. **Plot** `|Z_total|` vs. frequency from 50 Hz to 10 kHz.  
   At what frequency does `|Z_L|` equal `|Z_R|`? Mark it on the plot.  
   (This is the "corner frequency" $f_c = R / (2\pi L)$)

**Think about it:** The inductor doesn't dissipate heat — so where does the extra energy go?


```python
# ══════════════════════════════════════════════════════════════════════════════
# YOUR TURN 6 — Phone Charger Impedance
# ══════════════════════════════════════════════════════════════════════════════
#
# THE REAL-WORLD CONTEXT
# ───────────────────────
# Every phone charger contains an inductor (a coil of wire).
# Inductors slow down rapidly-changing currents (AC), but pass DC fine.
# They act as "speed bumps" — their opposition to current flow DEPENDS on
# the frequency of the alternating current.
#
# WHY CAN'T A REAL NUMBER DESCRIBE IMPEDANCE?
# ─────────────────────────────────────────────
# In a resistor:  current and voltage are ALWAYS IN STEP (in phase)
#   V = I * R    — real number R is enough
#
# In an inductor: current LAGS voltage by 90° (a quarter-cycle behind)
#   This phase shift CANNOT be expressed with a real number!
#   A 90° phase lag = multiplication by j (the imaginary unit)
#
# Impedance Z is a complex number:
#   • Real part  (resistance R):    energy is DISSIPATED as heat
#   • Imaginary part (reactance X): energy is STORED and RETURNED (no heat)
#
# Z_R = R + 0j   (pure resistor — no phase shift)
# Z_L = 0 + jωL  (pure inductor — 90° phase shift, ω = 2πf)
#
# A real number can't tell you about phase. Without phase information,
# you can't correctly compute power flow in AC circuits.

import numpy as np, matplotlib.pyplot as plt

R = 5         # resistance (ohms)
L = 2e-3      # inductance (henries = 2 mH, typical for a small charger coil)
V = 120 + 0j  # US wall voltage, 120 V RMS, taken as phase reference = 0°
f = 60        # frequency (Hz)
omega = 2 * np.pi * f   # angular frequency in rad/s

# ── Step 1: Individual impedances ────────────────────────────────────────
# Resistor: pure resistance, no phase shift → real number
Z_R = 5 + 0j
print(f"Z_R (resistor)  = {Z_R}  Ω")
print(f"  Phase angle = {np.degrees(np.angle(Z_R)):.1f}°  (zero — no phase shift)")

# Inductor: opposes AC proportional to frequency → purely imaginary
# Z_L = jωL  means: voltage LEADS current by 90° (or current lags voltage by 90°)
Z_L = 1j * omega * L
print(f"\nZ_L (inductor)  = {Z_L:.4f}  Ω")
print(f"  |Z_L| = {abs(Z_L):.4f} Ω  (opposition to 60 Hz current)")
print(f"  Phase angle = {np.degrees(np.angle(Z_L)):.1f}°  (current 90° behind voltage)")

# ── Step 2: Total series impedance ───────────────────────────────────────
# In series: impedances add like vectors (real parts sum, imaginary parts sum)
Z_total = Z_R + Z_L
print(f"\nZ_total = Z_R + Z_L = {Z_total:.4f}  Ω")
print(f"  |Z_total| = {abs(Z_total):.4f} Ω  (total opposition to current flow)")
print(f"  Phase angle = {np.degrees(np.angle(Z_total)):.2f}°  (current lags voltage by this angle)")
print(f"  This angle is between 0° (pure R) and 90° (pure L) — expected!")

# ── Step 3: Current (Ohm's Law in complex form) ────────────────────────────
# V = I * Z  →  I = V / Z
# Division of complex numbers automatically handles the phase shift
I = V / Z_total
print(f"\nCurrent I = V / Z = {I:.4f}  A")
print(f"  |I| = {abs(I):.4f} A  (amplitude)")
print(f"  Phase of I = {np.degrees(np.angle(I)):.2f}°  (lags behind V because of inductor)")
print(f"\n  Voltage phase:  0.00°  (reference)")
print(f"  Current phase: {np.degrees(np.angle(I)):.2f}°  (lags by this many degrees)")

# ── Step 4: Heat dissipated in resistor ───────────────────────────────────
# Only the resistor generates heat (the inductor stores and returns energy)
P_heat = abs(I)**2 * R
print(f"\nHeat dissipated in R = |I|² × R = {abs(I)**2:.4f} × {R} = {P_heat:.4f} W")

# Compare to resistor-only circuit (no inductor):
I_ideal = V / (R + 0j)
P_ideal = abs(I_ideal)**2 * R
print(f"Without inductor: P_heat = {P_ideal:.4f} W")
print(f"Inductor {'reduces' if P_heat < P_ideal else 'increases'} heat by {abs(P_heat-P_ideal):.4f} W")
print("(Inductor reduces current magnitude → less heat — it filters current without wasting energy)")

# ── Step 5: Plot |Z| vs frequency ─────────────────────────────────────────
# As frequency increases, the inductor fights back harder (|Z_L| = ωL grows)
# This is WHY phone chargers use inductors — to filter out high-frequency noise!
freqs   = np.linspace(50, 10000, 500)
Z_freq  = R + 1j * 2 * np.pi * freqs * L
f_corner = R / (2 * np.pi * L)   # frequency where |Z_R| = |Z_L|

print(f"\nCorner frequency f_c = R/(2πL) = {f_corner:.1f} Hz")
print(f"  Below f_c: inductor has little effect (nearly pure resistance)")
print(f"  Above f_c: inductor increasingly dominates (blocks high-frequency noise)")

plt.figure(figsize=(8, 4))
plt.plot(freqs, np.abs(Z_freq), 'b-', linewidth=2, label='|Z_total(f)|')
plt.axvline(f_corner, color='r', linestyle='--',
            label=f'Corner freq f_c = {f_corner:.0f} Hz')
plt.axhline(R, color='g', linestyle=':', label=f'R = {R} Ω (DC limit)')
plt.xlabel('Frequency (Hz)'); plt.ylabel('|Z| (Ω)')
plt.title('Charger Impedance vs Frequency\n(Higher frequency → higher impedance → better noise filtering)')
plt.legend(); plt.grid(True, alpha=0.3)
plt.tight_layout(); plt.show()
```

## 🏆 Challenge — Map Your Campus Walk

Using complex numbers as 2D positions (real = East, imaginary = North),  
represent a 5-stop walk around campus.

**You define the stops** — make up realistic building positions or use your real campus!  
Example stops: Dorm → Dining Hall → Library → Engineering Building → Classroom → back to Dorm.

**Required tasks:**
1. Create at least **5 stops** as complex numbers
2. Compute the **distance of each leg** (stop-to-stop)
3. Find the **total walk distance**
4. Find which **two stops are farthest apart** (check all pairs using a nested loop)
5. Plot the full route as a connected path with labeled dots
6. **Bonus:** Compute the "efficiency" of going directly from first to last stop  
   vs. the full route. What percentage of the path is "detour"?

This problem has no single right answer — the math works the same regardless of  
which campus you map!


```python
# ══════════════════════════════════════════════════════════════════════════════
# CHALLENGE — Full Campus Walk
# ══════════════════════════════════════════════════════════════════════════════
#
# This challenge combines EVERYTHING from the notebook into one real scenario:
#
#   Complex positions  →  campus map (like YT1)
#   Subtraction        →  displacement vectors between stops (like YT1)
#   abs()              →  walking distances (like YT2)
#   Path sum           →  total walk distance (like YT2)
#   cmath.phase()      →  compass bearings between stops
#   Efficiency ratio   →  how direct your route is (like YT2)
#   itertools          →  brute-force search over all stop pairs
#
# WHY A DICTIONARY OF COMPLEX NUMBERS?
# ──────────────────────────────────────
# A dict mapping name → complex position is essentially a labeled 2-D map.
# Each entry is a point on campus.  Operations on positions (distance,
# bearing, displacement) work directly on the complex values.
# No separate x[] and y[] arrays needed — the complex number is the point.

import cmath, math, itertools, matplotlib.pyplot as plt

# ── Step 1: Define campus stops ────────────────────────────────────────────
# Origin (0,0) = main campus gate; units = meters
# Real part = East (+) or West (-) of gate
# Imag part = North (+) or South (-) of gate
stops = {
    "Dorm":          -200 +  50j,   # 200 m West, 50 m North
    "Dining Hall":   -120 + 180j,   # 120 m West, 180 m North
    "Library":          0 + 250j,   # dead North, 250 m from gate
    "Eng Building":   180 + 150j,   # 180 m East, 150 m North
    "Classroom":      220 +  40j,   # 220 m East, 40 m North
}

stop_names = list(stops.keys())
stop_pos   = list(stops.values())

# Print the map
print("Campus stop positions:")
for name, pos in stops.items():
    print(f"  {name:15s}  East={pos.real:>6.0f} m   North={pos.imag:>5.0f} m")

# ── Step 2: Compute distances along the route ─────────────────────────────
print("\nRoute: " + " → ".join(stop_names))
print("Leg distances:")
total_dist = 0
for i in range(len(stop_pos) - 1):
    start, end = stop_pos[i], stop_pos[i+1]
    leg_vec  = end - start              # displacement vector for this leg
    leg_dist = abs(leg_vec)             # Euclidean length
    bearing  = math.degrees(cmath.phase(leg_vec))
    total_dist += leg_dist
    print(f"  {stop_names[i]:15s} → {stop_names[i+1]:15s}: "
          f"{leg_dist:>6.1f} m   bearing {bearing:>7.1f}°")

print(f"  {'TOTAL':>35}: {total_dist:>6.1f} m")

# ── Step 3: Farthest-apart stops (brute force) ────────────────────────────
print("\nSearching for the two stops farthest apart...")
max_dist = 0
max_pair = ("", "")
for (n1, p1), (n2, p2) in itertools.combinations(stops.items(), 2):
    d = abs(p2 - p1)
    if d > max_dist:
        max_dist = d
        max_pair = (n1, n2)
print(f"  Farthest pair: {max_pair[0]}  ↔  {max_pair[1]}  ({max_dist:.1f} m apart)")

# ── Step 4: Plot the route ────────────────────────────────────────────────
xs = [p.real for p in stop_pos]
ys = [p.imag for p in stop_pos]

fig, ax = plt.subplots(figsize=(9, 7))
ax.plot(xs, ys, 'b-o', linewidth=2.5, markersize=10, zorder=3)
for name, pos in stops.items():
    ax.annotate(name, xy=(pos.real, pos.imag),
                xytext=(pos.real + 8, pos.imag + 10),
                fontsize=10, fontweight='bold')
ax.plot(0, 0, 'r^', markersize=12, label='Main Gate (origin)', zorder=4)
ax.axhline(0, color='gray', lw=0.5)
ax.axvline(0, color='gray', lw=0.5)
ax.set_xlabel('East / West (meters)'); ax.set_ylabel('North / South (meters)')
ax.set_title('My Campus Walk — Complex Number Map')
ax.legend(); ax.grid(True, alpha=0.3)
plt.tight_layout(); plt.show()

# ── Step 5: Route efficiency ──────────────────────────────────────────────
direct = abs(stop_pos[-1] - stop_pos[0])   # straight-line start to finish
efficiency = direct / total_dist * 100
print(f"\nRoute efficiency analysis:")
print(f"  Total walking distance:        {total_dist:.1f} m")
print(f"  Direct Dorm → Classroom:       {direct:.1f} m")
print(f"  Routing efficiency:            {efficiency:.1f}%")
print(f"  Extra walking (detours):       {100-efficiency:.1f}% more than the direct path")
print(f"\nThis is the Triangle Inequality at work!")
print(f"|leg1| + |leg2| + |leg3| + |leg4|  ≥  |total displacement|")
print(f"{total_dist:.1f}  ≥  {direct:.1f}   ✓")

# ── Summary: What complex numbers let us do in this challenge ─────────────
print(f"\n{'='*55}")
print("What complex numbers contributed to this challenge:")
print("  • Stored 2-D positions in one variable (not two)")
print("  • Subtraction gave displacement vectors automatically")
print("  • abs() gave Euclidean distances without Pythagoras")
print("  • cmath.phase() gave compass bearings in one call")
print("  • All operations worked directly on the dict values")
print(f"{'='*55}")
```
