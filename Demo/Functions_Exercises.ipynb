## What is a Function?

A **function** is a reusable block of code that:
- Takes **inputs** (called parameters or arguments)
- Does some **work**
- Returns an **output** (or performs an action)

```python
def function_name(parameter1, parameter2):
    # do some work
    return result
```

**Why use functions?**
- Write the code once, use it many times
- Makes programs easier to read and debug
- Break big problems into smaller, manageable pieces

## Section 3 — Functions with Logical Conditions

Sometimes decisions require checking **multiple conditions at once** using `and`, `or`, and `not`.

````python
# ── EXAMPLE 3: Electrical circuit — check if in valid operating range ─────────
def circuit_status(voltage, current):
    """
    Checks if a circuit is operating within safe limits.
    Safe range: voltage 110-130 V AND current 0-15 A
    """
    voltage_ok  = (110 <= voltage <= 130)
    current_ok  = (0 <= current <= 15)

    if voltage_ok and current_ok:
        return "Operating normally"
    elif not voltage_ok and not current_ok:
        return "WARNING: Both voltage and current out of range"
    elif not voltage_ok:
        return f"WARNING: Voltage {voltage}V is out of range (110-130V)"
    else:
        return f"WARNING: Current {current}A is out of range (0-15A)"

# Test cases
print(circuit_status(120, 10))   # Normal
print(circuit_status(95,  10))   # Bad voltage
print(circuit_status(120, 20))   # Bad current
print(circuit_status(90,  25))   # Both bad
````

## Section 4 — Engineering Application Problems

Now let's combine everything: functions, parameters, return values, and decision statements to solve real engineering problems.

````python
# ── EXAMPLE 4: Wind chill temperature ────────────────────────────────────────
def wind_chill(temp_F, wind_mph):
    """
    Computes the wind chill temperature in Fahrenheit.
    Valid when temp <= 50°F and wind >= 3 mph.
    Formula: NWS wind chill index.
    """
    if temp_F > 50 or wind_mph < 3:
        return temp_F   # wind chill not applicable, return actual temp

    wc = (35.74
          + 0.6215 * temp_F
          - 35.75 * wind_mph**0.16
          + 0.4275 * temp_F * wind_mph**0.16)
    return round(wc, 1)

def wind_chill_danger(temp_F, wind_mph):
    """Returns a danger level based on wind chill."""
    wc = wind_chill(temp_F, wind_mph)

    if wc <= -60:
        return wc, "EXTREME DANGER — frostbite in under 2 minutes"
    elif wc <= -40:
        return wc, "DANGER — frostbite in 5-10 minutes"
    elif wc <= -20:
        return wc, "VERY COLD — frostbite in 30 minutes"
    elif wc <= 0:
        return wc, "COLD — dress in layers"
    else:
        return wc, "Cool — standard cold weather precautions"

# Test
scenarios = [(30, 10), (10, 30), (-10, 40), (20, 5)]
for temp, wind in scenarios:
    wc, danger = wind_chill_danger(temp, wind)
    print(f"Temp={temp:4}°F, Wind={wind:2}mph → Wind chill={wc:6}°F  |  {danger}")
````

### Key Rules to Remember

```python
# 1. Define BEFORE you call
def my_function():     # define first
    return 42

result = my_function() # then call

# 2. Indentation matters — everything inside the function is indented
def add(a, b):
    result = a + b     # inside function — indented
    return result      # inside function — indented

# 3. return ends the function immediately
def sign(x):
    if x > 0:
        return "positive"  # function stops here if x > 0
    return "non-positive"  # only reached if x <= 0

# 4. Call one function from another — avoids repeating code
def bmi_category(weight, height):
    bmi = compute_bmi(weight, height)  # call the other function!
    if bmi < 18.5:
        return bmi, "Underweight"
```
