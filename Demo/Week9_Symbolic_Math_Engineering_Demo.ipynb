## 1. The Power of Symbolic vs. Numerical Computation

The key difference: numerical tools give you *a number*. Symbolic tools give you *the formula*.

```python
# NUMERICAL approach — must pick a specific value
import numpy as np
x_num = 1.0
result_num = x_num/4 + x_num/3
print(f"Numerical: x/4 + x/3 = {result_num:.10f}  ← floating point, can't simplify further")
print()

# SYMBOLIC approach — x stays as a variable
x = symbols('x')
result_sym = x/4 + x/3
print("Symbolic: x/4 + x/3 =", simplify(result_sym), "← exact! Works for ANY value of x")
print()

# We can still get a number by substituting
print("Substituting x=1:", result_sym.subs(x, 1), "← exact fraction")
print("As decimal:     ", float(result_sym.subs(x, 1)))
```

```python
# More examples of exact symbolic results
print("sqrt(8) =", sqrt(8))          # 2*sqrt(2), not 2.828...
print("sin(pi/6) =", sin(pi/6))      # 1/2, not 0.4999...
print("(1/3)^2 =", Rational(1,3)**2) # 1/9, not 0.1111...

# Convert to float only when you need a final number
print("\nsqrt(8) as float:", float(sqrt(8)))
print("sqrt(8) to 30 decimal places:", sqrt(8).evalf(30))
```

## 3. Solving Equations — Engineering Applications

`solve()` returns exact symbolic solutions — all roots, in formula form.

```python
# Solve the quadratic formula from first principles
a, b, c = symbols('a b c')
x = symbols('x')

roots = solve(a*x**2 + b*x + c, x)
print("Quadratic roots:")
for i, r in enumerate(roots):
    print(f"  x_{i+1} =", r)
```

```python
# ENGINEERING EXAMPLE: Wall thickness of a pressurized vessel
# Hoop stress: sigma = P*r/t  →  find required wall thickness t

P, r, t, sigma = symbols('P r t sigma', positive=True)

# Define the equation (sigma = P*r/t)
stress_eq = Eq(sigma, P*r/t)
print("Equation:", stress_eq)

# Solve for wall thickness t
t_formula = solve(stress_eq, t)[0]
print("\nWall thickness formula: t =", t_formula)

# Apply to specific design: P=5 MPa, r=0.3 m, sigma_allow=200 MPa
t_required = t_formula.subs([(P, 5), (r, 0.3), (sigma, 200)])
print(f"\nFor P=5 MPa, r=0.3 m, sigma=200 MPa:")
print(f"  Required thickness t = {float(t_required):.4f} m = {float(t_required)*1000:.1f} mm")
```

```python
# ENGINEERING EXAMPLE: Node Voltage Method (circuit analysis)
# KCL at two nodes in a resistor network

V1, V2 = symbols('V1 V2')

# KCL at V1: (V1-10)/2 + V1/4 + (V1-V2)/3 = 0
# KCL at V2: (V2-V1)/3 + V2/6 = 2
eq1 = Eq((V1-10)/2 + V1/4 + (V1-V2)/3, 0)
eq2 = Eq((V2-V1)/3 + V2/6, 2)

print("KCL equations:")
print(" Node 1:", eq1)
print(" Node 2:", eq2)

solution = solve([eq1, eq2], [V1, V2])
print("\nExact solution:", solution)
print(f"V1 = {solution[V1]} = {float(solution[V1]):.4f} V")
print(f"V2 = {solution[V2]} = {float(solution[V2]):.4f} V")
```

## 4. Complex Numbers in Engineering

AC circuits, control systems, and vibration all rely on complex numbers. SymPy's `I` (= √-1) keeps results exact.

```python
# Basic complex number operations — exact results!
z = 3 + 4*I
print("z = 3 + 4j")
print("  |z| = Abs(z) =", Abs(z), "  (Pythagoras: sqrt(3²+4²) = 5, exact!)")
print("  Re(z) =", re(z))
print("  Im(z) =", im(z))
print("  arg(z) =", arg(z))  # phase angle
```

```python
# ENGINEERING EXAMPLE: RLC Series Circuit — Find Resonant Frequency
# Impedance: Z = R + j*(omega*L - 1/(omega*C))
# Resonance occurs when Im(Z) = 0

omega, R_val, L, C = symbols('omega R L C', positive=True)

# Total impedance
Z = R_val + I*(omega*L - 1/(omega*C))
print("Impedance Z =", Z)
print("Imaginary part of Z =", im(Z))

# Find resonant frequency (Im(Z) = 0)
resonance = solve(im(Z), omega)
omega_r_formula = resonance[0]
print("\nResonant frequency formula: omega_r =", omega_r_formula)
print("(This is the classic 1/sqrt(LC) result!)")

# Numerical example: L = 10 mH, C = 100 µF
omega_r = float(omega_r_formula.subs([(L, 0.01), (C, 100e-6)]))
f_r = omega_r / (2 * float(pi))
print(f"\nFor L=10 mH, C=100 µF:")
print(f"  omega_r = {omega_r:.1f} rad/s")
print(f"  f_r = {f_r:.2f} Hz")
```

## 5. Symbolic Differentiation

`diff(expr, var)` applies all calculus rules automatically — chain rule, product rule, partial derivatives.

```python
x = symbols('x')

# Basic derivatives
f1 = x**4 + 3*x**2 - 7*x
print("f(x) =", f1)
print("f'(x) =", diff(f1, x))
print("f''(x) =", diff(f1, x, 2))
print()

# Chain rule is automatic!
f2 = sin(x**2)
print("g(x) = sin(x²)")
print("g'(x) =", diff(f2, x), "  ← chain rule applied!")
print()

# Partial derivatives
x, y = symbols('x y')
h = x**2*y + y**3
print("h(x,y) = x²y + y³")
print("∂h/∂x =", diff(h, x))
print("∂h/∂y =", diff(h, y))
```

```python
# ENGINEERING EXAMPLE: Beam Deflection Maximum
# A simply-supported beam with uniform load w
# Deflection: y(x) = wx(L³ - 2Lx² + x³) / (24EI)
# Find where deflection is maximum

x, w, L, E, I_beam = symbols('x w L E I', positive=True)

y = w*x*(L**3 - 2*L*x**2 + x**3) / (24*E*I_beam)
print("Beam deflection y(x):")
print(y)

# dy/dx = 0 at maximum deflection
dy = diff(y, x)
print("\ndy/dx =")
print(factor(dy))

# Solve dy/dx = 0
x_max = solve(dy, x)
print("\nCritical points (dy/dx = 0):", x_max)
print("→ Maximum at x = L/2 (midspan) ✓")

# Maximum deflection value
y_max = simplify(y.subs(x, L/2))
print("\nMaximum deflection:")
print("y_max =", y_max)
print("\nThis is the classic 5wL⁴/(384EI) formula — derived symbolically!")
```

```python
# ENGINEERING EXAMPLE: Projectile Kinematics
# y(t) = v0*t - (1/2)*g*t²
# Derive velocity and acceleration, find time to peak

t, v0, g = symbols('t v0 g', positive=True)
y_proj = v0*t - Rational(1,2)*g*t**2

velocity = diff(y_proj, t)
acceleration = diff(y_proj, t, 2)

print("Position:     y(t) =", y_proj)
print("Velocity:    dy/dt =", velocity)
print("Acceleration: d²y/dt² =", acceleration)

t_peak = solve(velocity, t)[0]
print("\nTime to peak height (velocity=0): t =", t_peak)

# Maximum height
y_peak = simplify(y_proj.subs(t, t_peak))
print("Maximum height: h_max =", y_peak)
```

## 6. Symbolic Integration

`integrate(expr, var)` finds exact antiderivatives. `integrate(expr, (var, a, b))` for definite integrals.

```python
x, t, tau = symbols('x t tau', positive=True)

# Indefinite integrals — exact antiderivatives
print("∫ 3x² + 2x dx =", integrate(3*x**2 + 2*x, x))
print("∫ sin²(x) dx =", integrate(sin(x)**2, x))
print("∫ exp(-t/τ) dt =", integrate(exp(-t/tau), t))
print()

# Definite integrals
print("∫₀³ x² dx =", integrate(x**2, (x, 0, 3)), "(exact integer!)")

# Symbolic limits
a, b = symbols('a b')
print("∫ₐᵇ x² dx =", integrate(x**2, (x, a, b)))

# Improper integral!
x = symbols('x')
print("\n∫₋∞^∞ exp(-x²) dx =", integrate(exp(-x**2), (x, -oo, oo)))
print("This is the famous Gaussian integral = sqrt(π)!")
```

```python
# ENGINEERING EXAMPLE: Spring energy and work
# F(x) = k*x  →  W = ∫₀ᴸ F dx

x, k_spring, L = symbols('x k L', positive=True)

F = k_spring * x  # spring force (Hooke's law)
W = integrate(F, (x, 0, L))
print("Spring force: F(x) =", F)
print("Work done: W = ∫₀ᴸ F dx =", W)
print("(This is the classic ½kL² formula!)")

print("\nFor k=500 N/m, L=0.1 m:")
print("W =", float(W.subs([(k_spring, 500), (L, 0.1)])), "J")
```

```python
# ENGINEERING EXAMPLE: Center of mass of a tapered beam
# Density varies linearly: rho(x) = rho0*(1 + x/L)

x, rho0, L = symbols('x rho0 L', positive=True)

rho = rho0 * (1 + x/L)  # density profile

total_mass = integrate(rho, (x, 0, L))
first_moment = integrate(x * rho, (x, 0, L))
x_cm = simplify(first_moment / total_mass)

print("Density: rho(x) =", rho)
print("Total mass =", total_mass)
print("Center of mass x_cm =", x_cm)
print("(It's at 5/9 of the beam length — heavier end pulls it right!")
```

## 7. Solving ODEs — Differential Equations

`dsolve()` finds the exact general (or particular) solution to an ODE.

```python
# Basic ODE syntax
t = symbols('t')
y = Function('y')   # y is an UNKNOWN FUNCTION of t

# First-order: dy/dt + y = 0
ode1 = Eq(y(t).diff(t) + y(t), 0)
sol1 = dsolve(ode1, y(t))
print("ODE: dy/dt + y = 0")
print("Solution:", sol1)
print()

# Second-order: y'' + 4y = 0  with IC: y(0)=1, y'(0)=0
ode2 = Eq(y(t).diff(t, 2) + 4*y(t), 0)
sol2 = dsolve(ode2, y(t), ics={y(0): 1, y(t).diff(t).subs(t, 0): 0})
print("ODE: y'' + 4y = 0,  y(0)=1, y'(0)=0")
print("Solution:", sol2)
```

```python
# ENGINEERING EXAMPLE: RC Circuit Charging
# Governing equation: R*C*dV/dt + V = Vs
# Initial condition: V(0) = 0 (uncharged capacitor)

t = symbols('t', positive=True)
V = Function('V')
R_sym, C_sym, Vs = symbols('R C Vs', positive=True)

ode_rc = Eq(R_sym*C_sym*V(t).diff(t) + V(t), Vs)
sol_rc = dsolve(ode_rc, V(t), ics={V(0): 0})

print("RC circuit ODE:", ode_rc)
print("\nSolution:", sol_rc)
print("\nThis is Vs*(1 - exp(-t/RC)) — the classic capacitor charging formula!")

# Plot the charging curve
V_expr = sol_rc.rhs.subs([(R_sym, 1000), (C_sym, 1e-3), (Vs, 5)])
print("\nFor R=1kΩ, C=1mF, Vs=5V:")
print("V(t) =", V_expr)
print("V at t=1s:", float(V_expr.subs(t, 1)), "V")
print("V at t=5s:", float(V_expr.subs(t, 5)), "V  (≈ Vs = 5V)")
```

```python
# ENGINEERING EXAMPLE: Spring-Mass-Damper (Mechanical Vibrations)
# m*x'' + c*x' + k*x = 0
# Find critical damping condition

t = symbols('t')
x = Function('x')
m_sym, c_sym, k_sym = symbols('m c k', positive=True)

ode_smd = Eq(m_sym*x(t).diff(t,2) + c_sym*x(t).diff(t) + k_sym*x(t), 0)
print("Spring-Mass-Damper ODE:", ode_smd)

# Critical damping: c² = 4mk
c_crit = solve(c_sym**2 - 4*m_sym*k_sym, c_sym)
print("\nCritical damping coefficient c_crit =", c_crit[0])
print("(This is the classic c_cr = 2*sqrt(km) result!)")

# Numerical example: m=5 kg, k=200 N/m
c_num = float(c_crit[0].subs([(m_sym, 5), (k_sym, 200)]))
print(f"\nFor m=5 kg, k=200 N/m:")
print(f"  c_crit = {c_num:.2f} N·s/m")
```

## 8. Putting It All Together — Full Engineering Problem

**Suspension bridge cable sag:** A cable of length $L$ under horizontal tension $H$ and uniform load $w$ has the shape:
$$y(x) = \frac{w x (L-x)}{2H}$$

Find: (a) maximum sag, (b) slope at supports, (c) total arc length

```python
# Suspension cable analysis — all symbolic, one setup
x, L, w, H = symbols('x L w H', positive=True)

# Cable shape
y = w*x*(L - x) / (2*H)
print("Cable profile: y(x) =", y)

# (a) Maximum sag: find where dy/dx = 0
dy = diff(y, x)
x_max = solve(dy, x)[0]
y_max = simplify(y.subs(x, x_max))
print(f"\n(a) Maximum sag at x = {x_max}, y_max = {y_max}")

# (b) Slope at left support (x=0)
slope_left = dy.subs(x, 0)
print(f"\n(b) Slope at x=0: dy/dx = {slope_left}")

# (c) Arc length: s = ∫₀ᴸ sqrt(1 + (dy/dx)²) dx
# (For small sag approximation: s ≈ L + (1/2)∫(dy/dx)² dx)
ds_approx = Rational(1,2) * integrate(dy**2, (x, 0, L))
s_approx = L + ds_approx
print(f"\n(c) Approximate arc length: s ≈ {simplify(s_approx)}")

# Numerical example: L=100m, w=5000 N/m, H=500000 N
vals = [(L,100), (w,5000), (H,500000)]
print(f"\nNumerical example (L=100m, w=5000 N/m, H=500000 N):")
print(f"  Max sag = {float(y_max.subs(vals)):.2f} m")
print(f"  Slope at support = {float(slope_left.subs(vals)):.4f} rad = {float(slope_left.subs(vals))*180/float(pi):.2f}°")
```

## 🎯 Summary

| Task | SymPy command | Example |
|------|--------------|--------|
| Declare variable | `symbols('x')` | `x = symbols('x', positive=True)` |
| Simplify | `simplify(expr)` | `simplify((x²-1)/(x-1))` → `x+1` |
| Solve equation | `solve(eq, x)` | Returns list of roots |
| Differentiate | `diff(f, x, n)` | `diff(sin(x²), x)` → `2x·cos(x²)` |
| Integrate | `integrate(f, x)` | Indefinite or `(x, a, b)` for definite |
| Solve ODE | `dsolve(ode, y(t))` | Add `ics={}` for initial conditions |
| Substitute | `.subs(x, val)` | Chain: `.subs([(x,1),(y,2)])` |
| To float | `float(expr)` | Or `.evalf(n)` for n digits |

**Key rule:** Work symbolically throughout — substitute numbers at the very end.
