## 1. Basic Line Plots and Style Options

`plt.plot(x, y)` is the workhorse function. Style is controlled with a short string combining color, marker, and line style — or with explicit keyword arguments for more control.

```python
x = np.linspace(0, 10, 50)
y = np.sin(x)

plt.plot(x, y)
plt.title("Basic line plot")
plt.show()
```

Style shortcuts — color, marker, and line style combined into one string:

```python
fig, axs = plt.subplots(1, 3, figsize=(12, 3))
axs[0].plot(x, y, 'r--')       # red, dashed
axs[0].set_title("'r--'")
axs[1].plot(x, y, 'go-')        # green, circle markers, solid line
axs[1].set_title("'go-'")
axs[2].plot(x, y, 'b^:')        # blue, triangle markers, dotted line
axs[2].set_title("'b^:'")
plt.tight_layout()
plt.show()
```

Keyword arguments give more explicit control (useful when a style string gets too cryptic):

```python
plt.plot(x, y, color='purple', linestyle='-.', marker='s', markersize=4, linewidth=1.5, label='sin(x)')
plt.legend()
plt.show()
```

## 2. Formatting: Labels, Title, Legend, Grid

A complete, presentation-ready plot almost always needs all four of these.

```python
y1 = np.sin(x)
y2 = np.cos(x)

plt.plot(x, y1, label='sin(x)')
plt.plot(x, y2, label='cos(x)')
plt.xlabel('x')
plt.ylabel('y')
plt.title('sin and cos')
plt.legend()
plt.grid(True)
plt.show()
```

## 3. Multiple Plots: Subplots and Separate Figures

`plt.subplot(rows, cols, index)` puts several plots in one figure window. Calling `plt.figure()` again starts an entirely new, independent window.

```python
t = np.linspace(0, 2*np.pi, 100)

plt.subplot(2, 2, 1); plt.plot(t, np.sin(t));   plt.title('sin')
plt.subplot(2, 2, 2); plt.plot(t, np.cos(t));   plt.title('cos')
plt.subplot(2, 2, 3); plt.plot(t, np.tan(t));   plt.title('tan'); plt.ylim(-5, 5)
plt.subplot(2, 2, 4); plt.plot(t, np.exp(-t));  plt.title('exp(-t)')
plt.tight_layout()
plt.show()
```

## 4. Bar Charts

Vertical (`bar`), horizontal (`barh`), and grouped bar charts — useful for comparing categories rather than showing a continuous trend.

```python
categories = ['Steel', 'Aluminum', 'Titanium', 'Copper']
strength = [400, 310, 900, 210]   # MPa, illustrative values

plt.bar(categories, strength, color='steelblue')
plt.ylabel('Yield strength (MPa)')
plt.title('Material comparison')
plt.show()
```

```python
plt.barh(categories, strength, color='seagreen')
plt.xlabel('Yield strength (MPa)')
plt.title('Material comparison (horizontal)')
plt.show()
```

Grouped bars — comparing two data series side by side per category:

```python
labels = ['Q1', 'Q2', 'Q3', 'Q4']
product_a = [23, 31, 28, 35]
product_b = [19, 25, 30, 27]

x_pos = np.arange(len(labels))
width = 0.35

plt.bar(x_pos - width/2, product_a, width, label='Product A')
plt.bar(x_pos + width/2, product_b, width, label='Product B')
plt.xticks(x_pos, labels)
plt.ylabel('Units sold')
plt.legend()
plt.title('Quarterly sales comparison')
plt.show()
```

## 5. Pie Charts and Histograms

Pie charts show proportions of a whole; histograms show the distribution of a dataset.

```python
values = [45, 30, 15, 10]
labels = ['Product A', 'Product B', 'Product C', 'Product D']

plt.pie(values, labels=labels, autopct='%1.0f%%')
plt.title('Market share')
plt.show()
```

```python
np.random.seed(0)
data = np.random.normal(loc=70, scale=10, size=300)   # e.g., simulated test scores

plt.hist(data, bins=20, color='coral', edgecolor='black')
plt.xlabel('Score')
plt.ylabel('Frequency')
plt.title('Distribution of scores')
plt.show()
```

## 6. Three-Dimensional Plots

3D plotting needs `projection='3d'` when creating the axes. Three common types: line plots (a path through 3D space), scatter plots (individual 3D points), and surface plots (a function of two variables, $z=f(x,y)$).

```python
from mpl_toolkits.mplot3d import Axes3D

t = np.linspace(0, 6*np.pi, 300)
x3 = np.sin(t)
y3 = np.cos(t)
z3 = t

fig = plt.figure()
ax = fig.add_subplot(projection='3d')
ax.plot3D(x3, y3, z3)
ax.set_title('3D line plot (helix)')
ax.set_xlabel('x'); ax.set_ylabel('y'); ax.set_zlabel('z')
plt.show()
```

```python
np.random.seed(1)
xs = np.random.rand(50) * 10
ys = np.random.rand(50) * 10
zs = np.random.rand(50) * 10

fig = plt.figure()
ax = fig.add_subplot(projection='3d')
ax.scatter(xs, ys, zs, c='darkred')
ax.set_title('3D scatter plot')
plt.show()
```

```python
X = np.linspace(-3, 3, 40)
Y = np.linspace(-3, 3, 40)
X, Y = np.meshgrid(X, Y)
Z = np.sin(np.sqrt(X**2 + Y**2))

fig = plt.figure()
ax = fig.add_subplot(projection='3d')
ax.plot_surface(X, Y, Z, cmap='viridis')
ax.set_title('3D surface plot')
plt.show()
```

## Wrap-Up: A Multi-Panel Summary Figure

Combine several plot types into one figure using subplots: a line plot, a bar chart, and a pie chart, side by side.


```python
fig, axs = plt.subplots(1, 3, figsize=(13, 4))

t_final = np.linspace(0, 10, 100)
axs[0].plot(t_final, np.sin(t_final), 'b')
axs[0].set_title('Line plot')
axs[0].grid(True)

axs[1].bar(['A', 'B', 'C'], [12, 25, 18], color='teal')
axs[1].set_title('Bar chart')

axs[2].pie([40, 35, 25], labels=['X', 'Y', 'Z'], autopct='%1.0f%%')
axs[2].set_title('Pie chart')

plt.tight_layout()
plt.show()
```
