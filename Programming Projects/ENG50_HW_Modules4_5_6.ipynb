### Part A — Total Distance Traveled

Distance = integral of speed over time:
$$d = \int_0^{15} v(t)\, dt = \int_0^{15} 25 e^{-0.4t}\, dt$$

```python
# ── Problem 3A ─────────────────────────────────────────
k  = 0.4
v0 = 25.0

# Define the speed function
def speed(t):
    return v0 * np.exp(-k * t)

# Q1: Use quad to find the total distance from t=0 to t=15
distance, error = None, None   # YOUR CODE HERE: replace with quad call
print(f"Total distance:        {distance:.2f} m")
print(f"Estimated error:       {error:.1e}")

# Q2: Analytic check — the exact answer is v0/k * (1 - e^(-k*15))
distance_exact = v0 / k * (1 - np.exp(-k * 15))
print(f"Exact formula answer:  {distance_exact:.2f} m")
print(f"Match? {abs(distance - distance_exact) < 0.001}")

# Q3: Visualize — shade the area under the speed curve
t = np.linspace(0, 15, 400)
v = speed(t)

plt.figure(figsize=(7, 4))
plt.plot(t, v, 'b-', linewidth=2, label='v(t) = 25·e^(−0.4t)')
plt.fill_between(t, v, alpha=0.3, color='blue', label=f'Distance = {distance:.1f} m')
plt.xlabel('Time (s)')
plt.ylabel('Speed (m/s)')
plt.title('Distance = Area Under Speed Curve')
plt.grid(True)
plt.legend()
plt.show()
```
