## Real-World Scenario

When you throw a ball straight up, its height over time follows a quadratic equation:

$$h(t) = -16t^2 + v_0t + h_0$$

where $h_0$ is the starting height, $v_0$ is the initial upward velocity, and $t$ is **time**. To find out *when* the ball reaches a specific height — most commonly, when it lands ($h = 0$) — you set $h(t)$ equal to that height and solve for $t$. Setting $h(t) = 0$ turns the height formula into exactly the kind of equation your program solves: $at^2 + bt + c = 0$, where the unknown $x$ (or $t$, here) represents **time**, not height. Engineers solve equations like this constantly to find out when something reaches a certain height, position, or condition — including hitting the ground. Your program will solve the general quadratic equation so it can answer questions like that for any set of numbers.

## Background

For a quadratic equation $ax^2 + bx + c = 0$, the discriminant is:

$$D = b^2 - 4ac$$

and the roots are:

$$x = \frac{-b \pm \sqrt{D}}{2a}$$

- If $D > 0$: two real roots
- If $D = 0$: one real root (a repeated root)
- If $D < 0$: no real roots
