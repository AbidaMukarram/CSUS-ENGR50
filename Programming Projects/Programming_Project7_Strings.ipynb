## Real-World Scenario

Industrial facilities use **sensor networks** to monitor temperature, pressure, and flow in
real time. Each sensor transmits a data record — a string containing the sensor ID, location,
reading value, and status — that a control system must parse, validate, and log.

In this project you will write Python code that reads a sensor log file, parses each line
using string operations, generates formatted reports, and writes alert summaries back to disk.
This is exactly the kind of data-processing pipeline used in SCADA (Supervisory Control and
Data Acquisition) systems.


## Background

The file `sensor_log2.txt` contains 15 sensor readings, one per line, with
**whitespace-separated** columns:

```
sensor_id   location              timestamp                value   status
T-101       Boiler_Room_A         2024-03-15_08:00:12      87.4    NORMAL
P-205       Pump_Station_2        2024-03-15_08:00:15     142.8    WARNING
T-103       Heat_Exchanger_B      2024-03-15_08:10:44     112.7    ALERT
...
```

**Key Python string tools you will use:**

| Operation | Syntax |
|---|---|
| Split a string on whitespace | `line.split()` |
| Extract a substring (slice) | `s[start:end]` |
| Convert string to float | `float(s)` |
| Left-align in N characters | `f'{value:<N}'` |
| Right-align float, 2 dec. places | `f'{value:>9.2f}'` |
| Check start of string | `s.startswith('...')` |
| Check all characters are digits | `s.isdigit()` |
| Join a list of strings | `'\n'.join(list_of_strings)` |
